Todo lo que está debajo de la línea «<i>a partir de aquí es el motor</i>» puedes ignorarlo tranquilamente.</div></div>

```python
# ═══════════════════════════════════════════════════════════════════════════
#  🎛️  PANEL DE CONTROL — ejecuta esta celda UNA vez al abrir el notebook.
#
#  Todo lo que cambies aquí afecta a TODAS las gráficas del notebook.
#  Si nunca tocas nada, igual funciona: los valores por defecto ya están bien.
# ═══════════════════════════════════════════════════════════════════════════

# ───────────────────────────────────────────────────────────────────────────
#  1) ¿QUÉ TAMAÑO Y QUÉ FORMATO QUIERES EXPORTAR?
# ───────────────────────────────────────────────────────────────────────────
ANCHO_CM = 20        # Ancho de la gráfica en centímetros. 20 ≈ una diapositiva.
ALTO_CM  = 11        # Alto en centímetros. Más alto = barras más largas.

FORMATOS = ["svg"]   # Qué archivos generar. Opciones: "svg", "png", "pdf".
                     # Ejemplo para sacar los dos: FORMATOS = ["svg", "png"]

TEXTO_EDITABLE_EN_SVG = True
# True  → el texto del SVG se puede editar después en Illustrator / Figma / Canva.
# False → el texto se convierte en dibujo: NO se edita, pero se ve idéntico
#         en cualquier computador aunque no tenga la fuente instalada.

# ───────────────────────────────────────────────────────────────────────────
#  2) ¿DÓNDE QUIERES QUE QUEDEN LOS ARCHIVOS?
# ───────────────────────────────────────────────────────────────────────────
DESCARGAR_AUTOMATICAMENTE = True    # Baja el archivo a tu computador al ejecutar.
GUARDAR_EN_DRIVE          = False   # Ponlo en True para guardarlos en tu Drive.
CARPETA_DRIVE             = "MyDrive/Graficas Data Storytelling"

CARPETA_SALIDA = "graficas"         # Carpeta temporal dentro de Colab.

# ───────────────────────────────────────────────────────────────────────────
#  3) LA PALETA DE COLORES  (probada para daltonismo — cámbiala solo si tu
#     marca lo exige, y en ese caso cambia los códigos, no el orden)
# ───────────────────────────────────────────────────────────────────────────
AZUL    = "#2a78d6"    # Color 1 — el protagonista por defecto
NARANJA = "#eb6834"    # Color 2
AGUA    = "#1baf7a"    # Color 3
AMARILLO= "#eda100"    # Color 4
ROSA    = "#e87ba4"    # Color 5
VERDE   = "#008300"    # Color 6
VIOLETA = "#4a3aa7"    # Color 7
ROJO    = "#e34948"    # Color 8

COLORES = [AZUL, NARANJA, AGUA, AMARILLO, ROSA, VERDE, VIOLETA, ROJO]

ACENTO       = AZUL        # El color de "lo que importa"
GRIS_APAGADO = "#b8b7af"   # El color del "contexto" (lo que NO es la historia)

# Tintas de texto y del papel. No suelen tocarse.
TINTA   = "#0b0b0b"   # Títulos
TINTA_2 = "#52514e"   # Subtítulos y etiquetas
TINTA_3 = "#898781"   # Ejes y notas al pie
REJILLA = "#e1e0d9"   # Líneas guía de fondo
EJE     = "#c3c2b7"   # Línea del eje
FONDO   = "#fcfcfb"   # Color del papel de la gráfica


# ═══════════════════════════════════════════════════════════════════════════
#  A PARTIR DE AQUÍ ES "EL MOTOR". No necesitas entenderlo ni modificarlo.
#  Solo ejecútalo. (Aun así, cada bloque está explicado por si tienes curiosidad.)
# ═══════════════════════════════════════════════════════════════════════════

import os, sys, io, textwrap
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from matplotlib.lines import Line2D

# ¿Estamos dentro de Google Colab o en un Jupyter de nuestro computador?
# Lo detectamos para no ejecutar cosas que solo existen en Colab.
try:
    import google.colab            # noqa: F401
    EN_COLAB = True
except Exception:
    EN_COLAB = False

os.makedirs(CARPETA_SALIDA, exist_ok=True)


# ── Elegir la mejor fuente disponible ──────────────────────────────────────
# Buscamos la primera tipografía de la lista que exista en esta máquina.
def _fuente_disponible(candidatas):
    from matplotlib import font_manager
    instaladas = {f.name for f in font_manager.fontManager.ttflist}
    for c in candidatas:
        if c in instaladas:
            return c
    return "DejaVu Sans"

FUENTE = _fuente_disponible(
    ["Roboto", "Liberation Sans", "Helvetica", "Arial", "DejaVu Sans"])


# ── El tema visual ─────────────────────────────────────────────────────────
# "rcParams" es el panel de estilos global de matplotlib: definimos una sola
# vez cómo se ven ejes, rejilla, fuentes y colores, y todas las gráficas del
# notebook lo heredan. Es el equivalente a una plantilla de PowerPoint.
def aplicar_tema():
    plt.rcParams.update({
        "figure.figsize":   (ANCHO_CM / 2.54, ALTO_CM / 2.54),
        "figure.dpi":       110,          # nitidez en pantalla
        "savefig.dpi":      300,          # nitidez al exportar PNG
        "figure.facecolor": FONDO,
        "axes.facecolor":   FONDO,
        "savefig.facecolor": FONDO,
        "savefig.bbox":     "tight",      # no recorta títulos ni etiquetas
        "savefig.pad_inches": 0.28,
        "font.family":      "sans-serif",
        "font.sans-serif":  [FUENTE, "DejaVu Sans"],
        "font.size":        11,
        "text.color":       TINTA,
        "axes.labelcolor":  TINTA_2,
        "axes.labelsize":   11,
        "axes.edgecolor":   EJE,
        "axes.linewidth":   1.0,
        "axes.grid":        True,
        "axes.axisbelow":   True,         # la rejilla SIEMPRE detrás de los datos
        "axes.spines.top":   False,       # quitamos el marco: economía visual
        "axes.spines.right": False,
        "axes.spines.left":  False,
        "axes.prop_cycle":  plt.cycler(color=COLORES),
        "grid.color":       REJILLA,
        "grid.linewidth":   1.0,
        "grid.linestyle":   "-",          # nunca punteada: sería ruido visual
        "xtick.color":      TINTA_3,
        "ytick.color":      TINTA_3,
        "xtick.labelsize":  10,
        "ytick.labelsize":  10,
        "xtick.major.size": 0,
        "ytick.major.size": 0,
        "legend.frameon":   False,        # leyenda sin caja
        "legend.fontsize":  10.5,
        "lines.linewidth":  2.0,
        "lines.solid_capstyle": "round",
        "svg.fonttype":     "none" if TEXTO_EDITABLE_EN_SVG else "path",
        "pdf.fonttype":     42,
    })

aplicar_tema()


# ── lienzo(): crea la hoja en blanco ───────────────────────────────────────
def lienzo(ancho_cm=None, alto_cm=None):
    """Devuelve (fig, ax): la 'hoja' y el 'área de dibujo' de una gráfica."""
    a = (ancho_cm or ANCHO_CM) / 2.54
    b = (alto_cm or ALTO_CM) / 2.54
    fig, ax = plt.subplots(figsize=(a, b))
    return fig, ax


# ── titular(): título, subtítulo y fuente, al estilo de la prensa ──────────
def titular(ax, titulo, subtitulo="", fuente=""):
    """Pone el texto arriba a la izquierda (así lee el ojo) y la fuente abajo.

    El TÍTULO debe ser la conclusión, no la descripción.
    Mal:  "Ingresos por canal"        →  eso ya lo dice el eje.
    Bien: "Instagram duplica a Email" →  eso es una historia.
    """
    # Si la gráfica ya tiene leyenda, medimos cuánto ocupa para escribir el
    # título POR ENCIMA de ella y que nunca se pisen.
    base = 1.03
    leyenda_actual = ax.get_legend()
    if leyenda_actual is not None:
        try:
            ax.figure.canvas.draw()
            caja = leyenda_actual.get_window_extent().transformed(
                ax.transAxes.inverted())
            base = max(base, caja.y1 + 0.035)
        except Exception:
            base = 1.13

    y_sub = base
    y_tit = base + 0.075 if subtitulo else base
    ax.text(0, y_tit, titulo, transform=ax.transAxes, ha="left", va="bottom",
            fontsize=16.5, fontweight="bold", color=TINTA)
    if subtitulo:
        ax.text(0, y_sub, subtitulo, transform=ax.transAxes, ha="left", va="bottom",
                fontsize=11.5, color=TINTA_2)
    if fuente:
        ax._pie_de_fuente = fuente     # se dibuja al exportar, ya medido
    return ax


def _dibujar_fuente(fig, ax):
    """Coloca la nota de fuente justo debajo de todo lo demás, sin encimarse."""
    texto = getattr(ax, "_pie_de_fuente", None)
    if not texto:
        return
    fig.canvas.draw()
    try:
        caja = ax.get_tightbbox().transformed(fig.transFigure.inverted())
        x0, y0 = caja.x0, caja.y0
    except Exception:
        x0, y0 = 0.0, 0.0
    fig.text(x0, y0 - 0.075, texto, ha="left", va="top",
             fontsize=9, color=TINTA_3)
    ax._pie_de_fuente = None           # para no repetirla si se re-exporta


# ── Utilidades de formato ──────────────────────────────────────────────────
def formato_miles(ax, eje="y", decimales=0, sufijo="", prefijo=""):
    """Escribe los números como 1.250 en vez de 1250 (formato colombiano)."""
    def _fmt(v, _):
        s = f"{v:,.{decimales}f}"
        s = s.replace(",", "@").replace(".", ",").replace("@", ".")
        return f"{prefijo}{s}{sufijo}"
    destino = ax.yaxis if eje == "y" else ax.xaxis
    destino.set_major_formatter(mticker.FuncFormatter(_fmt))


def solo_rejilla(ax, direccion="horizontal"):
    """Deja únicamente las líneas guía necesarias. Menos tinta = más claridad."""
    ax.grid(False)
    if direccion in ("horizontal", "ambas"):
        ax.grid(True, axis="y", color=REJILLA, linewidth=1.0)
    if direccion in ("vertical", "ambas"):
        ax.grid(True, axis="x", color=REJILLA, linewidth=1.0)
    if direccion == "ninguna":
        ax.grid(False)
    ax.set_axisbelow(True)


def destacar(n, indice, color=None, apagado=None):
    """Devuelve una lista de colores donde SOLO 'indice' resalta.

    Es la técnica más poderosa del data storytelling: una barra en color,
    el resto en gris. El ojo va directo al dato que cuenta la historia.
    """
    color = color or ACENTO
    apagado = apagado or GRIS_APAGADO
    indices = indice if isinstance(indice, (list, tuple, set)) else [indice]
    return [color if i in indices else apagado for i in range(n)]


def etiquetar_barras(ax, contenedor, formato="{:,.0f}", separacion=4,
                     horizontal=False, color=None, negrita=False, solo=None):
    """Escribe el valor en la punta de cada barra.

    'solo' permite etiquetar únicamente algunas barras (p. ej. solo=[0]),
    que es lo recomendable: etiquetar TODO es ruido, no información.
    """
    color = color or TINTA_2
    for i, barra in enumerate(contenedor):
        if solo is not None and i not in solo:
            continue
        valor = barra.get_width() if horizontal else barra.get_height()
        texto = formato.format(valor).replace(",", "@").replace(".", ",").replace("@", ".")
        if horizontal:
            ax.annotate(texto,
                        xy=(barra.get_width(), barra.get_y() + barra.get_height() / 2),
                        xytext=(separacion, 0), textcoords="offset points",
                        ha="left", va="center", fontsize=10.5, color=color,
                        fontweight="bold" if negrita else "normal")
        else:
            ax.annotate(texto,
                        xy=(barra.get_x() + barra.get_width() / 2, barra.get_height()),
                        xytext=(0, separacion), textcoords="offset points",
                        ha="center", va="bottom", fontsize=10.5, color=color,
                        fontweight="bold" if negrita else "normal")


def leyenda(ax, arriba=True, columnas=None, handles=None, labels=None):
    """Leyenda sin caja, arriba a la izquierda, en una sola fila."""
    n = columnas or (len(labels) if labels else len(ax.get_legend_handles_labels()[0]))
    args = dict(loc="lower left", bbox_to_anchor=(0, 1.005), ncol=max(n, 1),
                frameon=False, handlelength=1.1, handletextpad=0.6,
                columnspacing=1.6, borderpad=0)
    if handles is not None:
        ax.legend(handles, labels, **args)
    else:
        ax.legend(**args)


# ── exportar(): la función que hace TODO el trabajo final ──────────────────
_EXPORTADAS = []

def exportar(fig, nombre, formatos=None, mostrar=True):
    """Guarda la gráfica, la descarga y la deja lista para usar.

    fig     → la figura que acabas de dibujar
    nombre  → cómo se llamará el archivo (sin extensión)
    """
    formatos = formatos or FORMATOS
    for ax in fig.axes:
        _dibujar_fuente(fig, ax)

    nombre = "".join(c for c in nombre if c.isalnum() or c in " _-").strip().replace(" ", "_")
    rutas = []
    for ext in formatos:
        ruta = os.path.join(CARPETA_SALIDA, f"{nombre}.{ext}")
        fig.savefig(ruta, format=ext, bbox_inches="tight", pad_inches=0.28,
                    facecolor=FONDO, transparent=False)
        rutas.append(ruta)
        _EXPORTADAS.append(ruta)

    # Copia a Google Drive (solo si lo pediste y estamos en Colab)
    if GUARDAR_EN_DRIVE and EN_COLAB:
        import shutil
        destino = os.path.join("/content/drive", CARPETA_DRIVE)
        try:
            os.makedirs(destino, exist_ok=True)
            for r in rutas:
                shutil.copy(r, destino)
        except Exception as e:
            print(f"⚠️  No se pudo escribir en Drive ({e}). "
                  f"¿Ejecutaste la celda de 'Conectar Google Drive'?")

    # Descarga automática al computador
    if DESCARGAR_AUTOMATICAMENTE and EN_COLAB:
        from google.colab import files
        for r in rutas:
            try:
                files.download(r)
            except Exception:
                print(f"⚠️  El navegador bloqueó la descarga de {r}. "
                      f"Búscalo en el panel 📁 de la izquierda.")

    print("✅ Listo:  " + "   ".join(rutas))
    if mostrar:
        plt.show()
    else:
        plt.close(fig)
    return rutas


print(f"✅ Panel de control cargado.")
print(f"   Entorno .......... {'Google Colab' if EN_COLAB else 'Jupyter local'}")
print(f"   Tipografía ....... {FUENTE}")
print(f"   Tamaño ........... {ANCHO_CM} x {ALTO_CM} cm")
print(f"   Formatos ......... {', '.join(FORMATOS)}")
print(f"   Carpeta .......... ./{CARPETA_SALIDA}/")
print("\n👉 Ya puedes ir a cualquier gráfica del catálogo y ejecutarla.")
```

```python
# ═══════════════════════════════════════════════════════════════════════════
#  📂  CARGAR TUS DATOS  (OPCIONAL)
#
#  Si NO ejecutas esta celda, el notebook funciona igual: cada gráfica trae
#  sus propios datos de ejemplo escritos a mano.
#  Ejecútala cuando quieras conectar tu propio archivo .csv o .xlsx.
# ═══════════════════════════════════════════════════════════════════════════

def cargar_datos(ruta=None):
    """Abre un CSV o Excel resolviendo solo los problemas típicos:
    separador (coma / punto y coma), tildes y ñ (encoding)."""
    import pandas as pd

    # Si no nos dan ruta y estamos en Colab, mostramos el botón para subir.
    if ruta is None:
        if not EN_COLAB:
            raise ValueError(
                "Fuera de Colab no hay botón para subir archivos. Escribe la ruta "
                "del tuyo, por ejemplo: cargar_datos('datos_ejemplo_marketing.csv')")
        from google.colab import files
        subidos = files.upload()                 # botón "Elegir archivos"
        ruta = list(subidos.keys())[0]

    # Excel: pandas lo resuelve directo.
    if ruta.lower().endswith((".xlsx", ".xls")):
        df = pd.read_excel(ruta)
    else:
        # CSV: probamos combinaciones hasta que una funcione.
        df = None
        for sep in [",", ";", "\t", "|"]:
            for enc in ["utf-8", "latin-1"]:
                try:
                    prueba = pd.read_csv(ruta, sep=sep, encoding=enc)
                    if prueba.shape[1] > 1:      # >1 columna = separador correcto
                        df = prueba
                        break
                except Exception:
                    continue
            if df is not None:
                break
        if df is None:
            raise ValueError("No pude leer el archivo. Revisa que sea un CSV válido.")

    # Limpieza mínima: quitamos espacios sobrantes en los nombres de columna.
    df.columns = [str(c).strip() for c in df.columns]

    print(f"✅ Datos cargados: {df.shape[0]:,} filas × {df.shape[1]} columnas")
    print(f"\n📋 Tus columnas (cópialas tal cual en las gráficas):")
    for c in df.columns:
        print(f"   • {c}")
    return df


# ───────────────────────────────────────────────────────────────────────────
#  👇 ELIGE UNA DE LAS TRES OPCIONES (borra el # de la que quieras usar)
# ───────────────────────────────────────────────────────────────────────────

# OPCIÓN A — Subir un archivo desde tu computador (lo más común):
datos = cargar_datos()

# OPCIÓN B — Un archivo que ya está en Google Drive:
# datos = cargar_datos("/content/drive/MyDrive/mis_datos.csv")

# OPCIÓN C — Un archivo publicado en internet (link directo al .csv):
# datos = cargar_datos("https://.../archivo.csv")

datos.head()      # muestra las primeras 5 filas para confirmar que quedó bien
```

## A · Comparar magnitudes

En barras eso <b>siempre</b> exagera las diferencias y se considera manipulación.</div></div><div style="flex:1 1 calc(50% - 7px);background:#eef4fd;border:1px solid #2a78d62e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#2a78d6;margin-bottom:7px;">&#128172;&nbsp;Mensaje que comunica</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">«Esta categoría es más grande que esta otra, y la diferencia es <i>de este tamaño</i>.»</div></div></div>

```python
# ═══════════════════════════════════════════════════════════════════════════
#  01 · BARRAS VERTICALES (COLUMNAS)
# ═══════════════════════════════════════════════════════════════════════════

# ─── 1) TUS DATOS ──────────────────────────────────────────────────────────
categorias = ["Instagram", "TikTok", "Google Ads", "Email", "Punto de venta"]
valores    = [3179, 3161, 2184, 1085, 3423]

# 👉 PARA USAR TU ARCHIVO: borra las 2 líneas de arriba, quita el # de las 2 de
#    abajo (solo el # y el espacio) y cambia los nombres de columna por los tuyos.
# resumen    = datos.groupby("canal")["ingresos"].sum().sort_values(ascending=False)
# categorias, valores = list(resumen.index), list(resumen.values)

# ─── 2) TUS TEXTOS ─────────────────────────────────────────────────────────
TITULO    = "El punto de venta físico aún manda"
SUBTITULO = "Ingresos acumulados 2025 por canal, en millones de pesos"
FUENTE    = "Fuente: base comercial de la compañía, ene–dic 2025."
EJE_Y     = "Millones COP"
RESALTAR  = 4        # posición de la barra protagonista (0 = la primera).
                     # Pon RESALTAR = None si no quieres resaltar ninguna.
ARCHIVO   = "01_barras_verticales"

# ─── 3) EL DIBUJO ──────────────────────────────────────────────────────────
fig, ax = lienzo()                       # hoja en blanco

# destacar() pinta de azul la barra que cuenta la historia y de gris el resto.
colores = destacar(len(valores), RESALTAR) if RESALTAR is not None else ACENTO

barras = ax.bar(categorias, valores,
                color=colores,
                width=0.55)              # 0.55 deja aire entre barras

ax.set_ylabel(EJE_Y)                     # nombre del eje vertical
formato_miles(ax, "y")                   # 3.179 en vez de 3179
solo_rejilla(ax, "horizontal")           # solo líneas guía horizontales

# Etiquetamos SOLO la barra protagonista: etiquetar todas sería ruido.
etiquetar_barras(ax, barras, solo=[RESALTAR] if RESALTAR is not None else None,
                 negrita=True, color=TINTA)

titular(ax, TITULO, SUBTITULO, FUENTE)

# ─── 4) EXPORTAR ───────────────────────────────────────────────────────────
exportar(fig, ARCHIVO)
```

Es la gráfica más segura y más legible que existe.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdefed;border:1px solid #c0392f2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#c0392f;margin-bottom:7px;">&#9940;&nbsp;No la uses si</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Las categorías son periodos de tiempo consecutivos: el tiempo se lee de izquierda a derecha, así que ahí van columnas o una línea.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdf6e6;border:1px solid #a9750a2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#a9750a;margin-bottom:7px;">&#9888;&#65039;&nbsp;Error t&iacute;pico</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">No ordenar de mayor a menor. Sin orden, el lector tiene que hacer el trabajo de comparar él solo, y no lo va a hacer.</div></div><div style="flex:1 1 calc(50% - 7px);background:#eef4fd;border:1px solid #2a78d62e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#2a78d6;margin-bottom:7px;">&#128172;&nbsp;Mensaje que comunica</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">«Este es el ranking, y estas son las distancias reales entre puestos.»</div></div></div>

<div style="background:#f0eefb;border:1px solid #4a3aa733;border-left:5px solid #4a3aa7;border-radius:11px;padding:15px 19px;color:#0b0b0b;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="font-size:14.5px;font-weight:800;color:#4a3aa7;margin-bottom:6px;">&#10024;&nbsp; El detalle que la hace ver profesional</div><div style="font-size:14.5px;line-height:1.65;color:#4a4945;">Al poner el valor al final de cada barra puedes <b>borrar el eje X entero</b>. Menos tinta, misma información. Es el principio de <i>economía visual</i> aplicado.</div></div>

```python
# ═══════════════════════════════════════════════════════════════════════════
#  02 · BARRAS HORIZONTALES
# ═══════════════════════════════════════════════════════════════════════════

# ─── 1) TUS DATOS ──────────────────────────────────────────────────────────
categorias = ["Regreso a clases", "Día de la Madre", "Black Friday", "Navidad"]
valores    = [1621, 1094, 1554, 3376]

# 👉 PARA USAR TU ARCHIVO:
# resumen    = datos[datos["campana"] != "Siempre activa"] \
#                  .groupby("campana")["ingresos"].sum()
# categorias, valores = list(resumen.index), list(resumen.values)

# ─── 2) TUS TEXTOS ─────────────────────────────────────────────────────────
TITULO    = "Navidad vende más que Black Friday y Día de la Madre juntos"
SUBTITULO = "Ingresos por campaña de temporada, 2025, en millones de pesos"
FUENTE    = "Fuente: base comercial de la compañía."
RESALTAR  = None      # aquí no resaltamos: el orden ya cuenta la historia
ARCHIVO   = "02_barras_horizontales"

# ─── 3) EL DIBUJO ──────────────────────────────────────────────────────────
# Regla de oro: en barras horizontales SIEMPRE se ordena de mayor a menor.
# Ordenamos las dos listas a la vez, de menor a mayor, porque matplotlib
# dibuja el eje Y de abajo hacia arriba.
pares = sorted(zip(valores, categorias))
valores    = [v for v, _ in pares]
categorias = [c for _, c in pares]

fig, ax = lienzo(alto_cm=0.9 * len(categorias) + 4)   # más categorías, más alto

colores = destacar(len(valores), RESALTAR) if RESALTAR is not None else ACENTO
barras  = ax.barh(categorias, valores, color=colores, height=0.6)

formato_miles(ax, "x")
solo_rejilla(ax, "vertical")

# En horizontales SÍ conviene etiquetar todas: no hay eje X que leer de golpe.
etiquetar_barras(ax, barras, horizontal=True)
ax.set_xticks([])                    # con las etiquetas al lado, el eje sobra
ax.grid(False)
ax.spines["bottom"].set_visible(False)
ax.margins(x=0.12)                   # aire a la derecha para que quepa el número

titular(ax, TITULO, SUBTITULO, FUENTE)

# ─── 4) EXPORTAR ───────────────────────────────────────────────────────────
exportar(fig, ARCHIVO)
```

### 03 · Barras agrupadas

<div style="display:flex;flex-wrap:wrap;gap:12px;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="flex:1 1 calc(50% - 7px);background:#ecf8f1;border:1px solid #0f8b4c2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#0f8b4c;margin-bottom:7px;">&#9989;&nbsp;&Uacute;sala cuando</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Comparas <b>2 o 3 series</b> dentro de varias categorías, y lo importante es comparar las series <i>entre sí</i> dentro de cada grupo.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdefed;border:1px solid #c0392f2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#c0392f;margin-bottom:7px;">&#9940;&nbsp;No la uses si</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Tienes más de 4 series, o lo que importa es el <b>total</b> de cada grupo. Para el total, usa apiladas (04).</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdf6e6;border:1px solid #a9750a2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#a9750a;margin-bottom:7px;">&#9888;&#65039;&nbsp;Error t&iacute;pico</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Meter 5, 6 o 7 series. La gráfica se vuelve una cebra ilegible y nadie compara nada.</div></div><div style="flex:1 1 calc(50% - 7px);background:#eef4fd;border:1px solid #2a78d62e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#2a78d6;margin-bottom:7px;">&#128172;&nbsp;Mensaje que comunica</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">«Dentro de cada categoría, así se ordenan estos grupos.»</div></div></div>

<div style="background:#fdf6e6;border:1px solid #a9750a33;border-left:5px solid #a9750a;border-radius:11px;padding:15px 19px;color:#0b0b0b;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="font-size:14.5px;font-weight:800;color:#a9750a;margin-bottom:6px;">&#9888;&#65039;&nbsp; El límite real son 3 series</div><div style="font-size:14.5px;line-height:1.65;color:#4a4945;">Con 4 ya cuesta, con 5 es imposible. Si necesitas más, divide en <b>varias gráficas pequeñas</b> (una por serie, con el mismo eje). Se llaman <i>small multiples</i> y siempre ganan.</div></div>

```python
# ═══════════════════════════════════════════════════════════════════════════
#  03 · BARRAS AGRUPADAS  (comparar varias series dentro de cada categoría)
# ═══════════════════════════════════════════════════════════════════════════

# ─── 1) TUS DATOS ──────────────────────────────────────────────────────────
categorias = ["Instagram", "TikTok", "Google Ads", "Email"]   # eje X

series = {                                   # cada llave = una barra de color
    "Bogotá":   [1148, 1141, 789, 392],
    "Medellín": [ 793,  789, 545, 271],
    "Cali":     [ 634,  631, 436, 217],
}

# 👉 PARA USAR TU ARCHIVO:
# tabla      = datos.pivot_table(index="canal", columns="region",
#                                values="ingresos", aggfunc="sum")
# categorias = list(tabla.index)
# series     = {col: list(tabla[col]) for col in tabla.columns}

# ─── 2) TUS TEXTOS ─────────────────────────────────────────────────────────
TITULO    = "Bogotá gana en todos los canales, pero la brecha se abre en redes"
SUBTITULO = "Ingresos 2025 por canal y ciudad, en millones de pesos"
FUENTE    = "Fuente: base comercial de la compañía."
ARCHIVO   = "03_barras_agrupadas"

# ─── 3) EL DIBUJO ──────────────────────────────────────────────────────────
# ⚠️ Máximo 3 o 4 series. Con más, el ojo ya no compara: usa 'small multiples'
#    (varias gráficas pequeñas, una por serie).
fig, ax = lienzo()

n       = len(series)
x       = np.arange(len(categorias))     # posiciones 0, 1, 2, 3...
ancho   = 0.72 / n                       # el 0.72 deja aire entre grupos

for i, (nombre, valores) in enumerate(series.items()):
    desplazamiento = (i - (n - 1) / 2) * ancho     # centra el grupo
    ax.bar(x + desplazamiento, valores, width=ancho * 0.9,
           label=nombre, color=COLORES[i])

ax.set_xticks(x)
ax.set_xticklabels(categorias)
ax.set_ylabel("Millones COP")
formato_miles(ax, "y")
solo_rejilla(ax, "horizontal")
leyenda(ax)                              # identificar series NUNCA por color solo

titular(ax, TITULO, SUBTITULO, FUENTE)

# ─── 4) EXPORTAR ───────────────────────────────────────────────────────────
exportar(fig, ARCHIVO)
```

### 04 · Barras apiladas

<div style="display:flex;flex-wrap:wrap;gap:12px;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="flex:1 1 calc(50% - 7px);background:#ecf8f1;border:1px solid #0f8b4c2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#0f8b4c;margin-bottom:7px;">&#9989;&nbsp;&Uacute;sala cuando</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Lo importante es el <b>total</b> de cada barra Y de qué se compone. Con <code>EN_PORCENTAJE = True</code> pasa a mostrar la <b>composición</b> (todas las barras suman 100 %).</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdefed;border:1px solid #c0392f2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#c0392f;margin-bottom:7px;">&#9940;&nbsp;No la uses si</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Quieres comparar con precisión los trozos <i>del medio</i> entre barras: no arrancan de la misma base, así que el ojo no puede.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdf6e6;border:1px solid #a9750a2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#a9750a;margin-bottom:7px;">&#9888;&#65039;&nbsp;Error t&iacute;pico</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Apilar 6 o 7 categorías. Solo el trozo de abajo y el total se leen bien; el resto es decoración.</div></div><div style="flex:1 1 calc(50% - 7px);background:#eef4fd;border:1px solid #2a78d62e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#2a78d6;margin-bottom:7px;">&#128172;&nbsp;Mensaje que comunica</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">«Este es el total, y este es el peso de cada parte dentro de él.»</div></div></div>

```python
# ═══════════════════════════════════════════════════════════════════════════
#  04 · BARRAS APILADAS  (parte-todo: de qué se compone cada barra)
# ═══════════════════════════════════════════════════════════════════════════

# ─── 1) TUS DATOS ──────────────────────────────────────────────────────────
categorias = ["T1", "T2", "T3", "T4"]        # eje X

series = {                                   # cada llave = un trozo de la barra
    "Punto de venta": [ 692,  838,  777, 1116],
    "Instagram":      [ 623,  770,  733, 1053],
    "TikTok":         [ 578,  749,  744, 1090],
    "Google Ads":     [ 447,  528,  494,  715],
}

# 👉 PARA USAR TU ARCHIVO:
# tabla      = datos.pivot_table(index="trimestre", columns="canal",
#                                values="ingresos", aggfunc="sum")
# categorias = list(tabla.index)
# series     = {col: list(tabla[col]) for col in tabla.columns}

# ─── 2) TUS TEXTOS ─────────────────────────────────────────────────────────
TITULO      = "El cuarto trimestre vale casi lo mismo que el primero y el segundo juntos"
SUBTITULO   = "Ingresos por trimestre y canal, en millones de pesos"
FUENTE      = "Fuente: base comercial de la compañía."
EN_PORCENTAJE = False    # ⬅ ponlo en True para ver la COMPOSICIÓN (barras 100%)
ARCHIVO     = "04_barras_apiladas"

# ─── 3) EL DIBUJO ──────────────────────────────────────────────────────────
fig, ax = lienzo()

matriz = np.array([series[k] for k in series], dtype=float)
if EN_PORCENTAJE:
    matriz = matriz / matriz.sum(axis=0) * 100      # cada barra suma 100

acumulado = np.zeros(len(categorias))
for i, nombre in enumerate(series):
    ax.bar(categorias, matriz[i], bottom=acumulado, width=0.55,
           label=nombre, color=COLORES[i],
           # El borde del color del papel crea el respiro de 2px entre trozos:
           edgecolor=FONDO, linewidth=1.6)
    # Etiquetamos dentro del trozo SOLO si el trozo es lo bastante grande.
    for j, valor in enumerate(matriz[i]):
        if valor / matriz[:, j].sum() > 0.12:       # más del 12% del total
            ax.text(j, acumulado[j] + valor / 2,
                    f"{valor:,.0f}".replace(",", ".") + ("%" if EN_PORCENTAJE else ""),
                    ha="center", va="center", fontsize=10,
                    color="#ffffff", fontweight="bold")
    acumulado += matriz[i]

ax.set_ylabel("% del trimestre" if EN_PORCENTAJE else "Millones COP")
if EN_PORCENTAJE:
    ax.set_ylim(0, 100)
formato_miles(ax, "y")
solo_rejilla(ax, "horizontal")
leyenda(ax)

titular(ax, TITULO, SUBTITULO, FUENTE)

# ─── 4) EXPORTAR ───────────────────────────────────────────────────────────
exportar(fig, ARCHIVO)
```

### 05 · Lollipop (paleta)

Ideal para tasas, promedios, índices y rankings largos.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdefed;border:1px solid #c0392f2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#c0392f;margin-bottom:7px;">&#9940;&nbsp;No la uses si</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Los valores son muy parecidos entre sí: el punto es pequeño y las diferencias mínimas se pierden.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdf6e6;border:1px solid #a9750a2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#a9750a;margin-bottom:7px;">&#9888;&#65039;&nbsp;Error t&iacute;pico</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Usarlo con valores negativos sin marcar el cero. El palito pierde su punto de partida.</div></div><div style="flex:1 1 calc(50% - 7px);background:#eef4fd;border:1px solid #2a78d62e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#2a78d6;margin-bottom:7px;">&#128172;&nbsp;Mensaje que comunica</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">«Este es el nivel de cada uno, y estos son los que se salen de la norma.»</div></div></div>

```python
# ═══════════════════════════════════════════════════════════════════════════
#  05 · LOLLIPOP  (la barra elegante: mismo mensaje, mucha menos tinta)
# ═══════════════════════════════════════════════════════════════════════════

# ─── 1) TUS DATOS ──────────────────────────────────────────────────────────
categorias = ["Email", "Punto de venta", "Google Ads", "Instagram", "TikTok"]
valores    = [9.52, 6.19, 3.79, 4.10, 3.38]     # ROAS: pesos por cada peso invertido

# 👉 PARA USAR TU ARCHIVO:
# resumen    = datos.groupby("canal")["roas"].mean()
# categorias, valores = list(resumen.index), list(resumen.values)

# ─── 2) TUS TEXTOS ─────────────────────────────────────────────────────────
TITULO    = "Email vende poco, pero es el canal más rentable"
SUBTITULO = "Retorno por peso invertido (ROAS) promedio por canal, 2025"
FUENTE    = "Fuente: base comercial de la compañía."
RESALTAR  = 4        # se calcula sobre la lista YA ORDENADA (0 = el de más arriba)
ARCHIVO   = "05_lollipop"

# ─── 3) EL DIBUJO ──────────────────────────────────────────────────────────
pares = sorted(zip(valores, categorias))
valores    = [v for v, _ in pares]
categorias = [c for _, c in pares]

fig, ax = lienzo(alto_cm=0.85 * len(categorias) + 4)

y = np.arange(len(categorias))
colores = destacar(len(valores), RESALTAR) if RESALTAR is not None else [ACENTO] * len(valores)

# El "palito": una línea fina desde 0 hasta el valor.
ax.hlines(y=y, xmin=0, xmax=valores, color=GRIS_APAGADO, linewidth=1.6, zorder=2)
# El "dulce": el punto al final, con anillo del color del papel.
ax.scatter(valores, y, s=190, color=colores, zorder=3,
           edgecolor=FONDO, linewidth=2)

for i, v in enumerate(valores):
    ax.annotate(f"{v:,.1f}".replace(".", ","), xy=(v, y[i]), xytext=(14, 0),
                textcoords="offset points", va="center", fontsize=10.5,
                color=TINTA if i == RESALTAR else TINTA_2,
                fontweight="bold" if i == RESALTAR else "normal")

ax.set_yticks(y)
ax.set_yticklabels(categorias)
ax.set_xlabel("Pesos ingresados por cada peso invertido")
ax.set_xlim(0, max(valores) * 1.18)
solo_rejilla(ax, "vertical")
ax.spines["bottom"].set_visible(False)

titular(ax, TITULO, SUBTITULO, FUENTE)

# ─── 4) EXPORTAR ───────────────────────────────────────────────────────────
exportar(fig, ARCHIVO)
```

## B · Evolución en el tiempo

<div style="background:#ecf8f1;border-radius:14px;padding:22px 26px;border-left:6px solid #0f8b4c;color:#0b0b0b;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:14px;flex-wrap:wrap;"><span style="display:inline-flex;align-items:center;justify-content:center;width:38px;height:38px;border-radius:11px;background:#0f8b4c;color:#ffffff;font-size:19px;font-weight:800;flex:0 0 auto;">B</span><span style="font-size:21px;font-weight:800;color:#0b0b0b;">Cuando la pregunta es «¿cómo viene cambiando?»</span></div><div style="font-size:15px;line-height:1.65;color:#4a4945;margin-top:12px;">La familia de las líneas. <div style="display:flex;flex-wrap:wrap;gap:12px;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="flex:1 1 calc(50% - 7px);background:#ecf8f1;border:1px solid #0f8b4c2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#0f8b4c;margin-bottom:7px;">&#9989;&nbsp;&Uacute;sala cuando</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Sigues <b>una sola serie</b> a lo largo del tiempo y quieres que se vea la tendencia, el punto de quiebre o la estacionalidad.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdefed;border:1px solid #c0392f2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#c0392f;margin-bottom:7px;">&#9940;&nbsp;No la uses si</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Tienes 3 o 4 puntos nada más: con tan pocos datos la línea sugiere una continuidad que no mediste. Por eso aquí el nombre va <b>al final de cada línea</b>.</div></div><div style="flex:1 1 calc(50% - 7px);background:#eef4fd;border:1px solid #2a78d62e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#2a78d6;margin-bottom:7px;">&#128172;&nbsp;Mensaje que comunica</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">«Estas dos venían iguales, y a partir de aquí se separaron.»</div></div></div>

<div style="background:#f0eefb;border:1px solid #4a3aa733;border-left:5px solid #4a3aa7;border-radius:11px;padding:15px 19px;color:#0b0b0b;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="font-size:14.5px;font-weight:800;color:#4a3aa7;margin-bottom:6px;">&#10024;&nbsp; El truco del énfasis también funciona en líneas</div><div style="font-size:14.5px;line-height:1.65;color:#4a4945;">Si tienes 8 series y solo una importa: píntala con <span style="background:#eceae4;color:#4a3aa7;border-radius:5px;padding:1.5px 6px;font-size:13px;font-family:ui-monospace,SFMono-Regular,Menlo,Consolas,monospace;">color=ACENTO</span> y todas las demás con <span style="background:#eceae4;color:#4a3aa7;border-radius:5px;padding:1.5px 6px;font-size:13px;font-family:ui-monospace,SFMono-Regular,Menlo,Consolas,monospace;">color=GRIS_APAGADO</span>. Sigues mostrando el contexto completo, pero el ojo va a la que cuenta la historia.</div></div>

```python
# ═══════════════════════════════════════════════════════════════════════════
#  07 · MULTI-LÍNEA  (comparar la evolución de varias series)
# ═══════════════════════════════════════════════════════════════════════════

# ─── 1) TUS DATOS ──────────────────────────────────────────────────────────
periodos = ["Ene", "Feb", "Mar", "Abr", "May", "Jun",
            "Jul", "Ago", "Sep", "Oct", "Nov", "Dic"]

series = {
    "Punto de venta": [222, 217, 249, 258, 297, 265, 254, 268, 277, 294, 379, 468],
    "Instagram":      [201, 197, 226, 234, 269, 240, 231, 244, 252, 268, 340, 392],
    "TikTok":         [166, 168, 197, 208, 244, 222, 217, 233, 245, 265, 352, 425],
}

# 👉 PARA USAR TU ARCHIVO:
# tabla    = datos.pivot_table(index="nombre_mes", columns="canal",
#                              values="ingresos", aggfunc="sum", sort=False) / 1e6
# periodos = list(tabla.index)
# series   = {col: list(tabla[col]) for col in tabla.columns}

# ─── 2) TUS TEXTOS ─────────────────────────────────────────────────────────
TITULO    = "TikTok adelantó a Instagram en el cierre del año"
SUBTITULO = "Ingresos mensuales por canal, en millones de pesos"
FUENTE    = "Fuente: base comercial de la compañía."
ARCHIVO   = "07_multilinea"

# ─── 3) EL DIBUJO ──────────────────────────────────────────────────────────
# ⚠️ Máximo 4 líneas. Con 5 o más se convierte en un plato de espaguetis:
#    usa 'small multiples' o destaca una sola y deja el resto en gris.
fig, ax = lienzo()

for i, (nombre, valores) in enumerate(series.items()):
    ax.plot(periodos, valores, color=COLORES[i], linewidth=2.2,
            label=nombre, zorder=3)
    # Punto + nombre al final de cada línea: se lee sin ir y volver a la leyenda.
    ax.scatter([len(periodos) - 1], [valores[-1]], s=80, color=COLORES[i],
               zorder=4, edgecolor=FONDO, linewidth=2)
    ax.annotate(f"{nombre}  {valores[-1]:,.0f}".replace(",", "."),
                xy=(len(periodos) - 1, valores[-1]), xytext=(11, 0),
                textcoords="offset points", va="center",
                fontsize=10.5, color=TINTA_2)

ax.set_ylabel("Millones COP")
formato_miles(ax, "y")
solo_rejilla(ax, "horizontal")
ax.margins(x=0.02)
ax.set_xlim(-0.4, len(periodos) + 3.4)     # aire a la derecha para los nombres
leyenda(ax)

titular(ax, TITULO, SUBTITULO, FUENTE)

# ─── 4) EXPORTAR ───────────────────────────────────────────────────────────
exportar(fig, ARCHIVO)
```

### 11 · Densidad

<div style="display:flex;flex-wrap:wrap;gap:12px;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="flex:1 1 calc(50% - 7px);background:#ecf8f1;border:1px solid #0f8b4c2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#0f8b4c;margin-bottom:7px;">&#9989;&nbsp;&Uacute;sala cuando</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Comparas la forma de <b>2 o 3 grupos</b> superpuestos. Es la versión suave y elegante del histograma, ideal para diapositivas.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdefed;border:1px solid #c0392f2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#c0392f;margin-bottom:7px;">&#9940;&nbsp;No la uses si</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Tienes pocos datos (menos de ~30 por grupo): la curva suaviza tanto que sugiere una forma que tus datos no respaldan.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdf6e6;border:1px solid #a9750a2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#a9750a;margin-bottom:7px;">&#9888;&#65039;&nbsp;Error t&iacute;pico</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Superponer 4, 5 o 6 curvas. Se convierte en una mancha de colores.</div></div><div style="flex:1 1 calc(50% - 7px);background:#eef4fd;border:1px solid #2a78d62e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#2a78d6;margin-bottom:7px;">&#128172;&nbsp;Mensaje que comunica</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">«Estos dos grupos se parecen / no se parecen, y así de distintos son.»</div></div></div>

```python
# ═══════════════════════════════════════════════════════════════════════════
#  11 · DENSIDAD  (el histograma suavizado: ideal para comparar 2 o 3 grupos)
# ═══════════════════════════════════════════════════════════════════════════
from scipy.stats import gaussian_kde

# ─── 1) TUS DATOS ──────────────────────────────────────────────────────────
np.random.seed(3)
grupos = {
    "Punto de venta": np.clip(np.random.normal(4.6, 0.35, 200), 1, 5),
    "Instagram":      np.clip(np.random.normal(4.2, 0.40, 200), 1, 5),
    "TikTok":         np.clip(np.random.normal(3.9, 0.45, 200), 1, 5),
}

# 👉 PARA USAR TU ARCHIVO:
# grupos = {nombre: grupo["satisfaccion"].dropna().values
#           for nombre, grupo in datos.groupby("canal")}

# ─── 2) TUS TEXTOS ─────────────────────────────────────────────────────────
TITULO    = "La tienda física es el canal que más satisface"
SUBTITULO = "Distribución de la satisfacción declarada (escala 1 a 5)"
FUENTE    = "Fuente: encuesta post-compra, n = 600."
EJE_X     = "Satisfacción (1 a 5)"
ARCHIVO   = "11_densidad"

# ─── 3) EL DIBUJO ──────────────────────────────────────────────────────────
# ⚠️ Máximo 3 curvas: más se enciman y dejan de leerse.
fig, ax = lienzo()

for i, (nombre, valores) in enumerate(grupos.items()):
    valores = np.asarray(valores, dtype=float)
    curva = gaussian_kde(valores)                       # suaviza el histograma
    eje_x = np.linspace(valores.min(), valores.max(), 300)
    eje_y = curva(eje_x)
    ax.fill_between(eje_x, eje_y, color=COLORES[i], alpha=0.13, zorder=2)
    ax.plot(eje_x, eje_y, color=COLORES[i], linewidth=2.2, label=nombre, zorder=3)

ax.set_xlabel(EJE_X)
ax.set_ylabel("Densidad")
ax.set_yticks([])                       # el número de densidad no dice nada útil
solo_rejilla(ax, "ninguna")
ax.spines["bottom"].set_visible(True)
leyenda(ax)

titular(ax, TITULO, SUBTITULO, FUENTE)

# ─── 4) EXPORTAR ───────────────────────────────────────────────────────────
exportar(fig, ARCHIVO)
```

### 12 · Dispersión (scatter)

<div style="display:flex;flex-wrap:wrap;gap:12px;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="flex:1 1 calc(50% - 7px);background:#ecf8f1;border:1px solid #0f8b4c2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#0f8b4c;margin-bottom:7px;">&#9989;&nbsp;&Uacute;sala cuando</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Quieres ver si dos variables numéricas se mueven juntas, y de paso detectar los casos que se salen de la norma.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdefed;border:1px solid #c0392f2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#c0392f;margin-bottom:7px;">&#9940;&nbsp;No la uses si</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Una de las dos variables es una categoría (canal, ciudad, producto). Es el error más caro en una presentación a directivos.</div></div><div style="flex:1 1 calc(50% - 7px);background:#eef4fd;border:1px solid #2a78d62e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#2a78d6;margin-bottom:7px;">&#128172;&nbsp;Mensaje que comunica</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">«A más de esto, más de aquello... o no.»</div></div></div>

<div style="background:#fdf6e6;border:1px solid #a9750a33;border-left:5px solid #a9750a;border-radius:11px;padding:15px 19px;color:#0b0b0b;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="font-size:14.5px;font-weight:800;color:#a9750a;margin-bottom:6px;">&#9888;&#65039;&nbsp; La frase que te puede costar el proyecto</div><div style="font-size:14.5px;line-height:1.65;color:#4a4945;">Correlación no es causalidad.
```python
# ═══════════════════════════════════════════════════════════════════════════
#  12 · DISPERSIÓN  (¿estas dos variables se mueven juntas?)
# ═══════════════════════════════════════════════════════════════════════════

# ─── 1) TUS DATOS ──────────────────────────────────────────────────────────
np.random.seed(5)
eje_x = np.random.uniform(4, 22, 120)                     # inversión
eje_y = eje_x * 4.2 + np.random.normal(0, 11, 120) + 8    # ingresos

# 👉 PARA USAR TU ARCHIVO:
# eje_x = datos["inversion"] / 1e6
# eje_y = datos["ingresos"]  / 1e6

# ─── 2) TUS TEXTOS ─────────────────────────────────────────────────────────
TITULO    = "A más inversión, más ingresos: la relación es casi una línea recta"
SUBTITULO = "Inversión vs. ingresos por campaña-mes-canal, en millones de pesos"
FUENTE    = "Fuente: base comercial de la compañía."
ETIQ_X    = "Inversión (millones COP)"
ETIQ_Y    = "Ingresos (millones COP)"
LINEA_TENDENCIA = True
ARCHIVO   = "12_dispersion"

# ─── 3) EL DIBUJO ──────────────────────────────────────────────────────────
fig, ax = lienzo()

# alpha=0.75 evita que los puntos amontonados se conviertan en una mancha.
ax.scatter(eje_x, eje_y, s=70, color=ACENTO, alpha=0.75,
           edgecolor=FONDO, linewidth=1.2, zorder=3)

if LINEA_TENDENCIA:
    # Recta de mejor ajuste: una sola línea que resume la relación.
    pendiente, corte = np.polyfit(eje_x, eje_y, 1)
    xs = np.linspace(min(eje_x), max(eje_x), 100)
    ax.plot(xs, pendiente * xs + corte, color=TINTA, linewidth=1.6,
            zorder=4, alpha=0.75)
    # La etiqueta va en la esquina superior IZQUIERDA, que aquí está vacía.
    ax.annotate(f"Cada millón invertido devuelve\n{pendiente:.1f} millones en ingresos".replace(".", ","),
                xy=(0.03, 0.94), xycoords="axes fraction",
                ha="left", va="top", fontsize=11, color=TINTA,
                fontweight="bold", linespacing=1.5)

ax.set_xlabel(ETIQ_X)
ax.set_ylabel(ETIQ_Y)
formato_miles(ax, "x"); formato_miles(ax, "y")
solo_rejilla(ax, "ambas")

titular(ax, TITULO, SUBTITULO, FUENTE)

# ─── 4) EXPORTAR ───────────────────────────────────────────────────────────
exportar(fig, ARCHIVO)
```

<div style="display:flex;flex-wrap:wrap;gap:12px;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="flex:1 1 calc(50% - 7px);background:#ecf8f1;border:1px solid #0f8b4c2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#0f8b4c;margin-bottom:7px;">&#9989;&nbsp;&Uacute;sala cuando</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Necesitas mostrar <b>tres</b> variables a la vez: dos en la posición y una en el tamaño. Clásico en matrices de portafolio y priorización de canales.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdefed;border:1px solid #c0392f2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#c0392f;margin-bottom:7px;">&#9940;&nbsp;No la uses si</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Las burbujas se encimarían mucho, o la tercera variable no aporta nada. En ese caso, dispersión simple (12).</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdf6e6;border:1px solid #a9750a2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#a9750a;margin-bottom:7px;">&#9888;&#65039;&nbsp;Error t&iacute;pico</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Escalar el <b>radio</b> en vez del <b>área</b>: duplica el valor y la burbuja se ve cuatro veces más grande. El código de abajo ya lo hace bien.</div></div><div style="flex:1 1 calc(50% - 7px);background:#eef4fd;border:1px solid #2a78d62e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#2a78d6;margin-bottom:7px;">&#128172;&nbsp;Mensaje que comunica</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">«Aquí están los grandes, aquí los rentables, y aquí el que es las dos cosas.»</div></div></div>

```python
# ═══════════════════════════════════════════════════════════════════════════
#  13 · BURBUJAS  (dispersión + una tercera variable en el tamaño)
# ═══════════════════════════════════════════════════════════════════════════

# ─── 1) TUS DATOS ──────────────────────────────────────────────────────────
etiquetas = ["Instagram", "TikTok", "Google Ads", "Email", "Punto de venta"]
eje_x     = [800, 944, 588, 115, 566]        # inversión del año (millones COP)
eje_y     = [4.0, 3.4, 3.8, 9.4, 6.1]        # ROAS (retorno por peso invertido)
tamano    = [3179, 3161, 2184, 1085, 3423]   # ingresos = tamaño de la burbuja
RESALTAR  = None                             # p. ej. RESALTAR = 3 destaca Email

# 👉 PARA USAR TU ARCHIVO:
# r = datos.groupby("canal").agg(x=("inversion","sum"), y=("roas","mean"),
#                                t=("ingresos","sum"))
# etiquetas = list(r.index)
# eje_x, eje_y, tamano = list(r.x / 1e6), list(r.y), list(r.t)

# ─── 2) TUS TEXTOS ─────────────────────────────────────────────────────────
TITULO    = "Email recibe la inversión más baja y devuelve el retorno más alto"
SUBTITULO = "Cada burbuja es un canal; su tamaño son los ingresos del año."
FUENTE    = "Fuente: base comercial de la compañía."
ETIQ_X    = "Inversión del año (millones COP)"
ETIQ_Y    = "ROAS (pesos ingresados por peso invertido)"
ARCHIVO   = "13_burbujas"

# ─── 3) EL DIBUJO ──────────────────────────────────────────────────────────
fig, ax = lienzo(alto_cm=12)

# El tamaño se escala por ÁREA, no por radio: si se escalara el radio, una
# burbuja del doble de valor se vería CUATRO veces más grande. Eso miente.
tam  = np.array(tamano, dtype=float)
area = 260 + (tam - tam.min()) / (tam.max() - tam.min() + 1e-9) * 1500

colores = destacar(len(eje_x), RESALTAR) if RESALTAR is not None else [ACENTO] * len(eje_x)

ax.scatter(eje_x, eje_y, s=area, color=colores,
           alpha=0.60, edgecolor=FONDO, linewidth=2, zorder=3)

# La etiqueta se separa según el TAMAÑO de su burbuja, para no montarse encima.
for i, nombre in enumerate(etiquetas):
    radio_pt = float(np.sqrt(area[i] / np.pi))
    ax.annotate(nombre, xy=(eje_x[i], eje_y[i]),
                xytext=(0, radio_pt + 9), textcoords="offset points",
                ha="center", va="bottom", fontsize=10.5,
                color=TINTA if i == RESALTAR else TINTA_2, fontweight="bold")

ax.set_xlabel(ETIQ_X)
ax.set_ylabel(ETIQ_Y)
solo_rejilla(ax, "ambas")
ax.margins(0.22)
ax.set_ylim(top=max(eje_y) * 1.16)      # aire arriba para la etiqueta más alta

titular(ax, TITULO, SUBTITULO, FUENTE)

# ─── 4) EXPORTAR ───────────────────────────────────────────────────────────
exportar(fig, ARCHIVO)
```

### 14 · Dona (donut)

<div style="display:flex;flex-wrap:wrap;gap:12px;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="flex:1 1 calc(50% - 7px);background:#ecf8f1;border:1px solid #0f8b4c2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#0f8b4c;margin-bottom:7px;">&#9989;&nbsp;&Uacute;sala cuando</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Tienes <b>máximo 5 porciones</b>, muy distintas entre sí, y el mensaje es «esta parte domina» o «está muy repartido». El agujero del centro sirve para poner el total.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdefed;border:1px solid #c0392f2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#c0392f;margin-bottom:7px;">&#9940;&nbsp;No la uses si</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Tienes más de 5 categorías, o dos porciones se parecen. Si necesitas que el lector <i>compare</i> las partes, la respuesta es barras horizontales (02).</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdf6e6;border:1px solid #a9750a2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#a9750a;margin-bottom:7px;">&#9888;&#65039;&nbsp;Error t&iacute;pico</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Ordenar las porciones al azar. Van siempre de mayor a menor, empezando arriba, en sentido horario.</div></div><div style="flex:1 1 calc(50% - 7px);background:#eef4fd;border:1px solid #2a78d62e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#2a78d6;margin-bottom:7px;">&#128172;&nbsp;Mensaje que comunica</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">«Un solo actor se lleva la mayor parte» o «no hay ningún actor dominante».</div></div></div>

<div style="background:#fdefed;border:1px solid #c0392f33;border-left:5px solid #c0392f;border-radius:11px;padding:15px 19px;color:#0b0b0b;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="font-size:14.5px;font-weight:800;color:#c0392f;margin-bottom:6px;">&#9940;&nbsp; Antes de usar esta gráfica, hazte una pregunta</div><div style="font-size:14.5px;line-height:1.65;color:#4a4945;">¿Puedo decir lo mismo con una barra? Casi siempre la respuesta es sí, y casi siempre la barra se entiende mejor. La dona se justifica cuando el mensaje es literalmente «<i>de un total</i>» y solo hay dos o tres protagonistas.</div></div>

```python
# ═══════════════════════════════════════════════════════════════════════════
#  14 · DONA  (parte-todo, y SOLO cuando son pocas porciones muy distintas)
# ═══════════════════════════════════════════════════════════════════════════

# ─── 1) TUS DATOS ──────────────────────────────────────────────────────────
etiquetas = ["Punto de venta", "Instagram", "TikTok", "Google Ads", "Email"]
valores   = [3423, 3179, 3161, 2184, 1085]

# 👉 PARA USAR TU ARCHIVO:
# resumen   = datos.groupby("canal")["ingresos"].sum().sort_values(ascending=False)
# etiquetas, valores = list(resumen.index), list(resumen.values)

# ─── 2) TUS TEXTOS ─────────────────────────────────────────────────────────
TITULO      = "Ningún canal supera el 26 % del ingreso"
SUBTITULO   = "Participación de cada canal en los ingresos de 2025"
FUENTE      = "Fuente: base comercial de la compañía."
CIFRA_CENTRO = "13.032"                 # el número grande del centro
TEXTO_CENTRO = "millones COP\nen total"
ARCHIVO     = "14_dona"

# ─── 3) EL DIBUJO ──────────────────────────────────────────────────────────
# ⚠️ Máximo 5 o 6 porciones. Si dos porciones se parecen, el ojo NO distingue
#    cuál es mayor: en ese caso usa barras horizontales (gráfica 02).
fig, ax = lienzo(ancho_cm=17, alto_cm=13)

total = sum(valores)
porciones, _ = ax.pie(
    valores,
    startangle=90,                       # arrancamos arriba, como un reloj
    counterclock=False,                  # y avanzamos en sentido horario
    colors=COLORES[:len(valores)],
    wedgeprops=dict(width=0.34,          # el "hueco" que la convierte en dona
                    edgecolor=FONDO, linewidth=2.5))

# Etiqueta fuera de cada porción: nombre + porcentaje.
for i, porcion in enumerate(porciones):
    angulo = np.deg2rad((porcion.theta1 + porcion.theta2) / 2)
    x, y = np.cos(angulo), np.sin(angulo)
    ax.annotate(f"{etiquetas[i]}\n{valores[i] / total * 100:.0f} %",
                xy=(x * 0.90, y * 0.90), xytext=(x * 1.22, y * 1.20),
                ha="center" if abs(x) < 0.4 else ("left" if x > 0 else "right"),
                va="center", fontsize=11, color=TINTA_2,
                linespacing=1.5)

# El agujero de la dona es espacio gratis: úsalo para el total.
ax.text(0, 0.10, CIFRA_CENTRO, ha="center", va="center",
        fontsize=27, fontweight="bold", color=TINTA)
ax.text(0, -0.16, TEXTO_CENTRO, ha="center", va="center",
        fontsize=11, color=TINTA_3, linespacing=1.5)

ax.set(aspect="equal")
ax.axis("off")
titular(ax, TITULO, SUBTITULO, FUENTE)

# ─── 4) EXPORTAR ───────────────────────────────────────────────────────────
exportar(fig, ARCHIVO)
```

### 15 · Ranking con énfasis y anotación

<div style="display:flex;flex-wrap:wrap;gap:12px;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="flex:1 1 calc(50% - 7px);background:#ecf8f1;border:1px solid #0f8b4c2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#0f8b4c;margin-bottom:7px;">&#9989;&nbsp;&Uacute;sala cuando</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Siempre que tengas <b>una conclusión concreta</b>. Es la gráfica que más aparece en presentaciones a directivos, y con razón.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdefed;border:1px solid #c0392f2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#c0392f;margin-bottom:7px;">&#9940;&nbsp;No la uses si</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Todavía estás explorando los datos y no sabes cuál es la historia. El énfasis exige haber decidido qué quieres decir.</div></div><div style="flex:1 1 calc(50% - 7px);background:#fdf6e6;border:1px solid #a9750a2e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#a9750a;margin-bottom:7px;">&#9888;&#65039;&nbsp;Error t&iacute;pico</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">Resaltar tres o cuatro barras «porque todas son importantes». Si resaltas varias, no resaltas ninguna.</div></div><div style="flex:1 1 calc(50% - 7px);background:#eef4fd;border:1px solid #2a78d62e;border-radius:11px;padding:14px 16px;color:#0b0b0b;min-width:250px;box-sizing:border-box;"><div style="font-size:11.5px;font-weight:800;letter-spacing:.09em;text-transform:uppercase;color:#2a78d6;margin-bottom:7px;">&#128172;&nbsp;Mensaje que comunica</div><div style="font-size:14px;line-height:1.6;color:#4a4945;">«De todo esto, lo que tienes que recordar es <i>esto</i>.»</div></div></div>

<div style="background:#ecf8f1;border:1px solid #0f8b4c33;border-left:5px solid #0f8b4c;border-radius:11px;padding:15px 19px;color:#0b0b0b;font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',Roboto,Helvetica,Arial,sans-serif;"><div style="font-size:14.5px;font-weight:800;color:#0f8b4c;margin-bottom:6px;">&#9989;&nbsp; Esta es la gráfica que deberías llevar a la sustentación</div><div style="font-size:14.5px;line-height:1.65;color:#4a4945;">Un jurado empresarial recuerda <b>una</b> idea por diapositiva. Esta gráfica está diseñada para que esa idea sea la que tú elegiste, y no la que el jurado dedujo por su cuenta.</div></div>

```python
# ═══════════════════════════════════════════════════════════════════════════
#  15 · RANKING CON ÉNFASIS  (la gráfica más útil del data storytelling)
# ═══════════════════════════════════════════════════════════════════════════

# ─── 1) TUS DATOS ──────────────────────────────────────────────────────────
etiquetas = ["Bogotá", "Medellín", "Cali", "Costa"]
valores   = [4318, 2977, 2382, 2085]

# 👉 PARA USAR TU ARCHIVO:
# resumen   = datos.groupby("region")["ingresos"].sum().sort_values()
# etiquetas, valores = list(resumen.index), list(resumen.values)

# ─── 2) TUS TEXTOS ─────────────────────────────────────────────────────────
TITULO      = "Bogotá aporta el doble que la Costa"
SUBTITULO   = "Ingresos 2025 por región, en millones de pesos"
FUENTE      = "Fuente: base comercial de la compañía."
PROTAGONISTA = "Bogotá"   # la categoría que cuenta la historia.
                          # Pon None para que resalte automáticamente la mayor.
ANOTACION   = "1 de cada 3 pesos\nde la compañía"
ARCHIVO     = "15_ranking_enfasis"

# ─── 3) EL DIBUJO ──────────────────────────────────────────────────────────
pares = sorted(zip(valores, etiquetas))
valores   = [v for v, _ in pares]
etiquetas = [c for _, c in pares]

fig, ax = lienzo(alto_cm=0.95 * len(etiquetas) + 4)

# Buscamos la categoría protagonista. Si no la encontramos (por una tilde, una
# mayúscula o un espacio de diferencia), resaltamos la mayor y avisamos.
if PROTAGONISTA in etiquetas:
    indice_protagonista = etiquetas.index(PROTAGONISTA)
else:
    indice_protagonista = len(valores) - 1        # la barra más larga
    if PROTAGONISTA is not None:
        print(f"⚠️  '{PROTAGONISTA}' no aparece en tus datos "
              f"(revisa tildes y mayúsculas). Resalto la mayor: "
              f"{etiquetas[indice_protagonista]}")
    PROTAGONISTA = etiquetas[indice_protagonista]

colores = destacar(len(valores), indice_protagonista)

barras = ax.barh(etiquetas, valores, color=colores, height=0.6)

# Solo la barra protagonista lleva su valor en negro; el resto en gris suave.
for i, barra in enumerate(barras):
    es_protagonista = (i == indice_protagonista)
    ax.annotate(f"{valores[i]:,.0f}".replace(",", "."),
                xy=(barra.get_width(), barra.get_y() + barra.get_height() / 2),
                xytext=(9, 0), textcoords="offset points", va="center",
                fontsize=11.5 if es_protagonista else 10.5,
                fontweight="bold" if es_protagonista else "normal",
                color=TINTA if es_protagonista else TINTA_3)

# La anotación: convierte el dato en frase. Esto es lo que recuerda la audiencia.
# La ponemos en el espacio VACÍO de la derecha, apuntando a la barra ganadora.
maximo = max(valores)
ax.annotate(ANOTACION,
            xy=(maximo * 0.97, indice_protagonista - 0.32),      # a dónde apunta
            xytext=(maximo * 0.82, indice_protagonista - 1.55),  # dónde va el texto
            ha="center", va="top", fontsize=11.5, color=ACENTO,
            fontweight="bold", linespacing=1.5,
            arrowprops=dict(arrowstyle="-|>", color=ACENTO, linewidth=1.5,
                            shrinkA=4, shrinkB=2,
                            connectionstyle="arc3,rad=-0.25"))

# Etiquetas del eje Y: la protagonista en negro, el resto en gris.
for etiqueta in ax.get_yticklabels():
    etiqueta.set_color(TINTA if etiqueta.get_text() == PROTAGONISTA else TINTA_3)
    etiqueta.set_fontweight("bold" if etiqueta.get_text() == PROTAGONISTA else "normal")

ax.set_xticks([])
ax.grid(False)
ax.spines["bottom"].set_visible(False)
ax.margins(x=0.14)

titular(ax, TITULO, SUBTITULO, FUENTE)

# ─── 4) EXPORTAR ───────────────────────────────────────────────────────────
exportar(fig, ARCHIVO)
```

```python
# ═══════════════════════════════════════════════════════════════════════════
#  16 · RECETARIO DE ANOTACIONES
#
#  Estas 5 recetas son COPIA Y PEGA: llévalas a cualquier gráfica de arriba,
#  justo antes de la línea exportar(...). Son lo que convierte un gráfico
#  en una historia.
# ═══════════════════════════════════════════════════════════════════════════

periodos = ["Ene", "Feb", "Mar", "Abr", "May", "Jun",
            "Jul", "Ago", "Sep", "Oct", "Nov", "Dic"]
valores  = [858, 838, 962, 993, 1150, 1027, 986, 1039, 1074, 1141, 1466, 1655]
META     = 1300

fig, ax = lienzo()
ax.plot(periodos, valores, color=ACENTO, linewidth=2.4, zorder=3)
ax.set_ylim(700, 2150)     # fijamos el alto ANTES de anotar, para colocar bien


# ── RECETA 1 · Línea de meta / promedio ────────────────────────────────────
# Da al lector la vara de medir: ¿esto es mucho o poco?
ax.axhline(META, color=TINTA_3, linewidth=1.3, zorder=2)
ax.annotate(f"Meta mensual: {META:,.0f}".replace(",", "."),
            xy=(0, META), xytext=(3, 6), textcoords="offset points",
            fontsize=10, color=TINTA_3)


# ── RECETA 2 · Sombrear un periodo ─────────────────────────────────────────
# Marca la temporada, la campaña o el evento que explica el cambio.
ax.axvspan(9.5, 11.5, color=AMARILLO, alpha=0.12, zorder=1)
ax.annotate("Temporada alta", xy=(10.5, 2105),
            ha="center", va="top", fontsize=10.5, color="#8a6000",
            fontweight="bold")


# ── RECETA 3 · Flecha que señala el dato clave ─────────────────────────────
ax.annotate("Black Friday empujó\nun 28 % sobre octubre",
            xy=(10, valores[10]),                     # punta de la flecha
            xytext=(7.4, 1930),                       # dónde va el texto
            ha="center", va="top",
            fontsize=11, color=TINTA, fontweight="bold", linespacing=1.5,
            arrowprops=dict(arrowstyle="-|>", color=TINTA,
                            linewidth=1.4, connectionstyle="arc3,rad=-0.25"))


# ── RECETA 4 · Resaltar un punto concreto ──────────────────────────────────
ax.scatter([11], [valores[11]], s=150, color=NARANJA, zorder=5,
           edgecolor=FONDO, linewidth=2.5)
ax.annotate(f"{valores[11]:,.0f}".replace(",", "."),
            xy=(11, valores[11]), xytext=(10, 4), textcoords="offset points",
            fontsize=12, fontweight="bold", color=NARANJA)


# ── RECETA 5 · Caja de "así se lee esta gráfica" ───────────────────────────
ax.text(0.02, 0.94,
        "Cada punto es el ingreso total del mes,\nsumando los cinco canales.",
        transform=ax.transAxes, va="top", ha="left", fontsize=10,
        color=TINTA_2, linespacing=1.6,
        bbox=dict(boxstyle="round,pad=0.6", facecolor="#eef4fd",
                  edgecolor="#c9dcf7", linewidth=1))


ax.set_ylabel("Millones COP")
formato_miles(ax, "y")
solo_rejilla(ax, "horizontal")
ax.margins(x=0.06)

titular(ax, "Cómo anotar una gráfica para que cuente una historia",
        "Las cinco recetas aplicadas a la vez (en la vida real, usa 1 o 2)",
        "Fuente: base comercial de la compañía.")

exportar(fig, "16_recetario_anotaciones")
```
