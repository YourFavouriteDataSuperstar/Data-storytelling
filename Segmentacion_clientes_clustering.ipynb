## El problema

La directora de marketing lo dice así:

> *«No quiero otra campaña para todos. Quiero saber **qué tipos de clientes tenemos**, cuánto vale cada tipo y a quién le hablo primero con el próximo presupuesto.»*

Fíjate en lo que **no** está pidiendo: no quiere predecir una variable que ya conoce (eso sería aprendizaje *supervisado*). Quiere **descubrir grupos que nadie ha nombrado todavía**. Eso es un problema de **segmentación**, y se resuelve con **aprendizaje no supervisado**.

## La pregunta de negocio, bien planteada

# Aprender sin respuestas: qué hace un modelo de agrupamiento

En el aprendizaje **supervisado** el modelo aprende con ejemplos que ya traen la respuesta («este cliente sí compró, este no»). En el **no supervisado** no hay respuesta: el modelo solo ve las características de cada cliente y **busca quiénes se parecen entre sí**. La palabra clave es **distancia**: dos clientes son parecidos si sus números están cerca.

Vamos a comparar dos formas muy distintas de agrupar:

<div style="display:flex;gap:12px;flex-wrap:wrap;margin:12px 0 16px;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><div style="flex:1;min-width:180px;border-top:4px solid #A3195B;background:#FBE8F1;padding:12px 14px;border-radius:0 0 8px 8px;"><div style="font-weight:700;color:#16201F;margin-bottom:4px;">Modelo A · K-means</div><div style="font-size:10.5pt;line-height:1.5;color:#2A2933;">Tú dices cuántos grupos quieres (<b>k</b>). El modelo pone k «centros» al azar, asigna cada cliente al centro más cercano, mueve cada centro al promedio de su grupo y repite hasta que nada cambia. Rápido y simple.</div></div><div style="flex:1;min-width:180px;border-top:4px solid #1B7E9C;background:#E1F0F5;padding:12px 14px;border-radius:0 0 8px 8px;"><div style="font-weight:700;color:#16201F;margin-bottom:4px;">Modelo B · Jerárquico (Ward)</div><div style="font-size:10.5pt;line-height:1.5;color:#2A2933;">Empieza con cada cliente solo y va <b>fusionando</b> los dos grupos más parecidos, uno a la vez, hasta que todos quedan juntos. El resultado es un árbol (dendrograma) que tú cortas a la altura que quieras.</div></div></div>

<div style="border-left:3px solid #C97A2B;background:#FBF0E4;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#9A5A16;">OJO</span><div style="font-weight:700;color:#16201F;margin:2px 0 4px;">K-means no es lo mismo que KNN</div><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">En clase y en internet vas a ver que se confunden <b>K-means</b> y <b>KNN</b> (<i>k-nearest neighbors</i>, k vecinos más cercanos). Se parecen en el nombre y en que los dos usan distancias, pero hacen cosas diferentes:<br><br>• <b>K-means</b> es <b>no supervisado</b>: <i>descubre</i> grupos que no existían. La k es el número de grupos.<br>• <b>KNN</b> es <b>supervisado</b>: <i>clasifica</i> un caso nuevo mirando a qué grupo pertenecen sus k vecinos más cercanos. Necesita que los grupos ya existan. La k es el número de vecinos.<br><br>Por eso en este cuaderno los usamos en orden: primero <b>descubrimos</b> los segmentos con K-means y con el jerárquico, y al final usamos <b>KNN</b> para asignar automáticamente a los clientes nuevos que lleguen mañana.</div></div>

| | K-means | Jerárquico (Ward) | KNN |
|---|---|---|---|
| Tipo | No supervisado | No supervisado | Supervisado |
| Para qué | Descubrir grupos | Descubrir grupos (y su jerarquía) | Asignar casos nuevos a grupos conocidos |
| Qué decides tú | Número de grupos *k* | Dónde cortar el árbol | Número de vecinos *k* |
| Lo usamos en | Secciones 6, 8–9 | Secciones 7–9 | Sección 10 |

## ¿Por qué hay que escalar?

Los dos modelos agrupan por **distancia**. Si comparamos dos clientes que difieren en 10.000 dólares de ingreso y en 2 hijos, la distancia la decide casi por completo el ingreso, porque 10.000 es un número mucho más grande que 2. El modelo «creería» que los hijos no importan.

La solución es **estandarizar**: a cada variable se le resta su promedio y se divide por su desviación estándar. Después de eso, todas quedan en la misma escala (promedio 0, desviación 1) y pesan lo mismo. Al gasto, además, le aplicamos un **logaritmo** para «desinflar» a los pocos clientes que gastan muchísimo y que, si no, dominarían el modelo.

```python
X = df[VARIABLES].copy()
X["Gasto"] = np.log1p(X["Gasto"])        # log(1 + gasto): suaviza la cola larga

escalador = StandardScaler()
Z = escalador.fit_transform(X)            # Z es la tabla que ven los modelos

comparacion = pd.DataFrame({"Antes · promedio": X.mean(), "Antes · desviación": X.std(),
                            "Después · promedio": Z.mean(axis=0), "Después · desviación": Z.std(axis=0)})
tabla(comparacion, formato="{:,.2f}")
```

# K-means: cómo funciona y cuántos grupos buscar

## Cómo funciona por dentro

K-means busca **k clientes típicos**, que se llaman *centroides*, y reparte a cada cliente con el típico que más se le parece. Lo hace con una receta de cuatro pasos:

<div style="display:flex;gap:12px;flex-wrap:wrap;margin:12px 0 16px;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><div style="flex:1;min-width:180px;border-top:4px solid #A3195B;background:#FBE8F1;padding:12px 14px;border-radius:0 0 8px 8px;"><div style="font-weight:700;color:#16201F;margin-bottom:4px;">1 · Elegir k</div><div style="font-size:10.5pt;line-height:1.5;color:#2A2933;">Tú decides cuántos grupos buscar. El algoritmo no lo sabe por su cuenta.</div></div><div style="flex:1;min-width:180px;border-top:4px solid #1B7E9C;background:#E1F0F5;padding:12px 14px;border-radius:0 0 8px 8px;"><div style="font-weight:700;color:#16201F;margin-bottom:4px;">2 · Arrancar</div><div style="font-size:10.5pt;line-height:1.5;color:#2A2933;">Pone k centros iniciales. scikit-learn usa <b>k-means++</b>: el primero al azar y los siguientes lejos de los ya elegidos, para no arrancar amontonado.</div></div><div style="flex:1;min-width:180px;border-top:4px solid #E8943A;background:#FCE9D3;padding:12px 14px;border-radius:0 0 8px 8px;"><div style="font-weight:700;color:#16201F;margin-bottom:4px;">3 · Asignar</div><div style="font-size:10.5pt;line-height:1.5;color:#2A2933;">Cada cliente se va con el centro más cercano. Así quedan k grupos.</div></div><div style="flex:1;min-width:180px;border-top:4px solid #8E5BD9;background:#F0E7FA;padding:12px 14px;border-radius:0 0 8px 8px;"><div style="font-weight:700;color:#16201F;margin-bottom:4px;">4 · Recalcular</div><div style="font-size:10.5pt;line-height:1.5;color:#2A2933;">Cada centro se mueve al <b>promedio</b> de sus clientes (de ahí el «means»). Se repiten los pasos 3 y 4 hasta que los centros dejan de moverse.</div></div></div>

**Qué intenta lograr.** Que la **inercia** sea lo más pequeña posible: la suma, para todos los clientes, de la distancia al cuadrado entre el cliente y su centro. Cada vuelta de asignar y recalcular la baja o la deja igual; cuando ya no baja, el algoritmo para.

**Qué distancia usa.** La **euclidiana**: la línea recta entre dos clientes en el espacio de las seis variables. Es el teorema de Pitágoras del colegio, con seis lados en vez de dos. Por eso escalamos antes: con variables en escalas distintas, la línea recta la decidiría una sola variable.

**Por qué se repite con varios arranques.** El resultado depende de dónde empiezan los centros. Un mal arranque puede quedarse en una solución buena pero no la mejor (un *mínimo local*). Con `n_init=20` se prueban 20 arranques y se queda el de menor inercia.

| Lo que K-means hace bien | Lo que hay que vigilar |
|---|---|
| Es rápido incluso con millones de clientes | Hay que decirle cuántos grupos buscar |
| Es fácil de explicar: cada segmento tiene un cliente típico | Supone grupos más o menos redondos y de tamaño parecido |
| Asigna clientes nuevos directo al centro más cercano | Los clientes atípicos jalan los centros (por eso limpiamos y usamos el logaritmo del gasto) |
| Con la semilla fija, todos obtienen el mismo resultado | Siempre entrega k grupos, aunque los datos no tengan grupos reales |

<div style="border-left:3px solid #E8943A;background:#FCE9D3;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#B26212;">PARA EL NEGOCIO</span><div style="font-weight:700;color:#16201F;margin:2px 0 4px;">El centroide es la persona del segmento</div><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">El centroide es un cliente que no existe, pero que resume al segmento: por ejemplo, «ingreso de 78 mil, gasta unos 1.300 en dos años, casi nunca compra con descuento». Es la base de la <i>persona</i> con la que marketing escribe los mensajes.</div></div>

Míralo en acción con un ejemplo pequeño: 300 puntos en dos dimensiones y tres centros que arrancan mal ubicados a propósito, los tres metidos en el mismo grupo.

```python
P, _ = make_blobs(n_samples=300, centers=[[-4, -2], [0, 3], [4, -1]], cluster_std=1.1, random_state=7)
centros = np.array([[-5.0, -3.0], [-3.0, -2.5], [-4.0, -0.5]])   # arranque malo: los tres en el grupo de la izquierda
colores_demo = [MAGENTA, TEAL, AMBAR]

historia = []
for paso in range(30):
    asignacion = np.argmin(((P[:, None, :] - centros[None]) ** 2).sum(axis=2), axis=1)   # paso 3: asignar
    historia.append((centros.copy(), asignacion))
    nuevos_centros = np.array([P[asignacion == j].mean(axis=0) for j in range(3)])    # paso 4: recalcular
    if np.allclose(nuevos_centros, centros):
        break
    centros = nuevos_centros

mostrar = sorted({0, 1, 2, len(historia) // 2, len(historia) - 1})[-4:] if len(historia) > 4 else list(range(len(historia)))
fig, ejes = plt.subplots(1, len(mostrar), figsize=(14, 4.2), sharex=True, sharey=True)
for ax, i in zip(ejes, mostrar):
    c, a = historia[i]
    for j in range(3):
        ax.scatter(P[a == j, 0], P[a == j, 1], s=12, color=colores_demo[j], alpha=0.55, linewidths=0)
    ax.scatter(c[:, 0], c[:, 1], marker="X", s=260, color=TINTA, edgecolor="white", linewidth=1.8, zorder=3)
    inercia = ((P - c[a]) ** 2).sum()
    ax.set_title("Paso 0 · arranque" if i == 0 else (f"Paso {i} · ya no se mueve" if i == mostrar[-1] else f"Paso {i}"))
    ax.text(0.03, 0.96, f"inercia {miles(inercia)}", transform=ax.transAxes, fontsize=10.5, color=TINTA, va="top",
            fontweight="semibold", bbox=dict(fc="white", ec="none", alpha=0.85))
    ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
fig.tight_layout(w_pad=1.5)
encabezado(fig, "K-means · ejemplo", f"Aunque arranque mal, K-means encuentra los tres grupos en {len(historia) - 1} pasos",
           "Las X son los centros. En cada paso los puntos se van con el centro más cercano, los centros se mueven al promedio de sus puntos y la inercia baja.",
           fuente="Datos simulados para el ejemplo")
plt.show()
```

## Elegir cuántos grupos

K-means necesita que le digamos **cuántos grupos (k)** buscar. Para elegir *k* se usan dos métricas, una al lado de la otra:

<div style="display:flex;gap:12px;flex-wrap:wrap;margin:12px 0 16px;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><div style="flex:1;min-width:180px;border-top:4px solid #A3195B;background:#FBE8F1;padding:12px 14px;border-radius:0 0 8px 8px;"><div style="font-weight:700;color:#16201F;margin-bottom:4px;">Inercia (método del codo)</div><div style="font-size:10.5pt;line-height:1.5;color:#2A2933;">Suma de las distancias de cada cliente a su centro. Siempre baja al aumentar k (con un grupo por cliente sería 0). Se busca el <b>codo</b>: el punto donde agregar un grupo más ya no mejora mucho.</div></div><div style="flex:1;min-width:180px;border-top:4px solid #E8943A;background:#FCE9D3;padding:12px 14px;border-radius:0 0 8px 8px;"><div style="font-weight:700;color:#16201F;margin-bottom:4px;">Silueta</div><div style="font-size:10.5pt;line-height:1.5;color:#2A2933;">Para cada cliente compara qué tan cerca está de su grupo frente al grupo vecino. Va de −1 a 1: cerca de 1, bien ubicado; cerca de 0, en la frontera; negativo, probablemente en el grupo equivocado. Se busca la <b>más alta</b>.</div></div></div>

Probamos de 2 a 8 grupos. Cada prueba corre K-means **20 veces** con puntos de partida distintos (`n_init=20`) y se queda con la mejor, porque K-means empieza al azar. La semilla `random_state=42` hace que el azar sea repetible: todos en clase verán los mismos resultados.

```python
pruebas = []
for k in range(2, 9):
    modelo = KMeans(n_clusters=k, n_init=20, random_state=42).fit(Z)
    pruebas.append({"k": k, "inercia": modelo.inertia_, "silueta": silhouette_score(Z, modelo.labels_)})
pruebas = pd.DataFrame(pruebas)

K_ELEGIDO = 4
fig, (a1, a2) = plt.subplots(1, 2, figsize=(12, 5))
a1.plot(pruebas["k"], pruebas["inercia"], color=GRIS, marker="o", linewidth=2)
fila = pruebas.set_index("k").loc[K_ELEGIDO]
a1.scatter([K_ELEGIDO], [fila["inercia"]], s=140, color=MAGENTA, zorder=3)
a1.set_title("Inercia: el codo se suaviza entre 3 y 4 grupos")
a1.set_xlabel("Número de grupos (k)"); a1.set_ylabel("Inercia (menos es mejor)")

colores = [MAGENTA if k == K_ELEGIDO else (AMBAR if k == pruebas.loc[pruebas.silueta.idxmax(), "k"] else GRIS_CLARO)
           for k in pruebas["k"]]
a2.bar(pruebas["k"], pruebas["silueta"], color=colores, width=0.6)
for k, s in zip(pruebas["k"], pruebas["silueta"]):
    a2.text(k, s + 0.006, f"{s:.2f}", ha="center", fontsize=10, color=TINTA)
a2.set_title("Silueta: la más alta es k = 2")
a2.set_xlabel("Número de grupos (k)"); a2.set_ylabel("Silueta promedio (más es mejor)")
a2.set_ylim(0, pruebas["silueta"].max() * 1.18)
fig.tight_layout(w_pad=3)
encabezado(fig, "K-means · Fig. 3", "Las métricas prefieren 2 grupos; el negocio necesita 4",
           "Con k = 2 la silueta es la más alta, pero solo separa «gastan mucho» de «gastan poco», algo que marketing ya sabe. Con k = 4 la silueta baja poco y aparecen grupos con los que sí se puede actuar.")
plt.show()
```

# Clustering jerárquico: el árbol genealógico de los clientes

El jerárquico no necesita que le digas *k* desde el principio. Funciona de abajo hacia arriba:

1. Cada cliente empieza siendo su propio grupo (2.212 grupos).
2. Se fusionan los dos grupos más parecidos.
3. Se repite hasta que solo queda un grupo con todos.

Usamos el criterio de **Ward**: en cada paso fusiona los dos grupos cuya unión hace crecer **lo menos posible** la dispersión interna. Es el que produce grupos más compactos y comparables con K-means.

El resultado se dibuja como un **dendrograma**: un árbol donde la altura de cada unión indica qué tan distintos eran los grupos que se juntaron. **Cortar el árbol** con una línea horizontal define cuántos segmentos te quedan.

## Cómo se decide qué tan parecidos son dos grupos

Entre dos clientes la distancia es clara: la línea recta. Entre dos **grupos** de clientes hay que escoger una regla, que se llama **enlace** (*linkage*). La regla cambia por completo el resultado:

| Enlace | Mide la distancia entre dos grupos como… | Qué tipo de grupos produce |
|---|---|---|
| Simple | la de sus dos miembros más cercanos | Cadenas largas: un cliente «puente» puede unir dos grupos muy distintos |
| Completo | la de sus dos miembros más lejanos | Grupos compactos, pero muy sensibles a los atípicos |
| Promedio | el promedio de todas las parejas posibles | Un punto medio entre las dos anteriores |
| **Ward** (el que usamos) | cuánto crecería la dispersión interna si se juntan | Grupos compactos y de tamaño parecido, comparables con los de K-means |

| Lo que el jerárquico hace bien | Lo que hay que vigilar |
|---|---|
| No hay que decidir el número de grupos antes: el árbol muestra todas las opciones | Cada fusión es definitiva: una decisión temprana no se corrige después |
| Con los mismos datos siempre da el mismo resultado, sin azar | Con muchos clientes se vuelve lento y pesado, porque compara todas las parejas |
| Muestra segmentos y subsegmentos anidados | No tiene una forma propia de asignar clientes nuevos |

Primero, un ejemplo con diez clientes para ver cómo se arma el árbol.

```python
puntos = np.array([[1.0, 1.0], [1.6, 1.3], [1.2, 2.0], [5.0, 5.0], [5.6, 5.4], [4.8, 6.0],
                   [9.0, 1.2], [9.6, 1.6], [8.8, 2.1], [5.3, 2.6]])
letras = list("ABCDEFGHIJ")
arbol_demo = linkage(puntos, method="ward")
set_link_color_palette([MAGENTA, TEAL, AMBAR])
corte_demo = (arbol_demo[-3, 2] + arbol_demo[-2, 2]) / 2

fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 5), gridspec_kw={"width_ratios": [1, 1.3]})
a1.scatter(puntos[:, 0], puntos[:, 1], s=260, color="white", edgecolor=TINTA, linewidth=1.5, zorder=3)
for (x, y), letra in zip(puntos, letras):
    a1.text(x, y, letra, ha="center", va="center", fontsize=10, fontweight="bold", color=TINTA, zorder=4)
a1.set_title("Diez clientes en dos variables"); a1.set_xticks([]); a1.set_yticks([])
a1.set_xlim(0, 10.5); a1.set_ylim(0, 7)
dendrogram(arbol_demo, labels=letras, color_threshold=corte_demo, above_threshold_color=GRIS_CLARO, ax=a2)
a2.axhline(corte_demo, color=MAGENTA, linestyle="--", linewidth=1.3)
a2.set_title("Su dendrograma"); a2.set_ylabel("Distancia de fusión"); a2.grid(axis="x", visible=False)
fig.tight_layout(w_pad=3)
encabezado(fig, "Jerárquico · ejemplo", "El árbol junta primero a los vecinos y deja para el final a los grupos lejanos",
           "Las uniones bajas juntan clientes casi idénticos; las altas, grupos muy distintos. Se corta donde hay un salto grande de altura.",
           fuente="Datos simulados para el ejemplo")
plt.show()
```

## Ponerles nombre a los grupos

Para comparar los dos modelos necesitamos hablar el mismo idioma. Esta función mira el perfil promedio de cada grupo y le asigna un nombre de negocio con reglas simples y explícitas:

* El grupo que **más gasta** → **Premium**.
* Entre los demás, el de **menor ingreso** → **Jóvenes en exploración** (ya veremos por qué «jóvenes» y por qué «exploración»).
* De los dos que quedan, el que **más gasta** → **Familias consolidadas**; el otro → **Familias cazadoras de ofertas**.

Así, si en tu computador el modelo numera los grupos en otro orden, los nombres siguen saliendo bien.

```python
def nombrar_segmentos(etiquetas):
    perfil = df.assign(g=etiquetas).groupby("g")[["Gasto", "Ingreso"]].mean()
    nombres = {perfil["Gasto"].idxmax(): "Premium"}
    resto = perfil.drop(index=list(nombres))
    nombres[resto["Ingreso"].idxmin()] = "Jóvenes en exploración"
    resto = resto.drop(index=resto["Ingreso"].idxmin())
    nombres[resto["Gasto"].idxmax()] = "Familias consolidadas"
    nombres[resto["Gasto"].idxmin()] = "Familias cazadoras de ofertas"
    return pd.Series(etiquetas, index=df.index).map(nombres)

df["seg_km"] = pd.Categorical(nombrar_segmentos(df["grupo_km"]), ORDEN_SEGMENTOS, ordered=True)
df["seg_jer"] = pd.Categorical(nombrar_segmentos(df["grupo_jer"]), ORDEN_SEGMENTOS, ordered=True)

tabla(pd.DataFrame({"K-means": df["seg_km"].value_counts(), "Jerárquico": df["seg_jer"].value_counts()})
      .loc[ORDEN_SEGMENTOS])
```

# ¿Qué modelo agrupa mejor? Las métricas, una por una

En un modelo supervisado medimos aciertos contra la respuesta correcta. Aquí no hay respuesta correcta, así que medimos **qué tan buenos son los grupos** desde varios ángulos. Cada métrica responde una pregunta distinta, y por eso se usan juntas. Este es el mapa:

| Métrica | Pregunta que responde | Rango | Mejor si es… | Sirve para |
|---|---|---|---|---|
| **Inercia** | ¿Qué tan apretados están los clientes alrededor de su centro? | 0 a ∞ | Baja, buscando el codo | Elegir k dentro de K-means |
| **Silueta** | ¿Cada cliente está más cerca de su grupo que del vecino? | −1 a 1 | **Alta** | Comparar modelos y valores de k |
| **Calinski-Harabasz** | ¿Los grupos son compactos y lejanos entre sí? | 0 a ∞ | **Alta** | Comparar modelos sobre los mismos datos |
| **Davies-Bouldin** | ¿Hay grupos que se confunden con su vecino? | 0 a ∞ | **Baja** | Comparar modelos sobre los mismos datos |
| **Estabilidad (ARI)** | ¿Salen los mismos segmentos si cambian un poco los datos? | 0 a 1 | **Alta** | Saber si el modelo aguanta mes a mes |
| **Lift** | ¿Los segmentos responden distinto en algo que el modelo no vio? | 0 a ∞ | Que se separen de 1 | Conectar el modelo con el negocio |

Las cuatro primeras son métricas **internas**: solo miran la forma de los grupos. La estabilidad mira si el resultado es **confiable**. El lift es una validación **externa**: compara contra el negocio.

## Cómo comparar dos algoritmos sin hacer trampa

1. **Mismos datos, mismas variables, misma escala.** Si cambia cualquiera de las tres, las métricas no son comparables.
2. **Mismo número de grupos.** Comparar K-means con 4 contra el jerárquico con 6 mezcla dos decisiones distintas.
3. **Varias métricas a la vez.** Cada una premia una forma de grupo. Silueta y Calinski-Harabasz favorecen los grupos redondos que busca K-means; si solo miras esas, el juego está inclinado.
4. **La estabilidad, siempre.** Un modelo que cambia con los datos no sirve para operar.
5. **Una validación externa.** Algo que el modelo no vio: aquí, las campañas.
6. **Al final, el negocio.** Tamaño de los segmentos, si cambian una decisión y cuánto cuesta actuar sobre ellos.

Antes de calcularlas, dos figuras para ver las métricas por dentro. Primero, la silueta de **un solo cliente** en un ejemplo pequeño:

```python
P2, g2 = make_blobs(n_samples=90, centers=[[0, 0], [4.2, 0], [2.1, 3.6]], cluster_std=0.75, random_state=3)
dist = np.linalg.norm(P2[:, None] - P2[None], axis=2)
propio = 0
del_grupo = np.where(g2 == propio)[0]
i = del_grupo[np.argmax(P2[del_grupo, 0])]          # el cliente del grupo 0 que está más a la derecha
a = dist[i, (g2 == propio) & (np.arange(len(P2)) != i)].mean()
b_por_grupo = {g: dist[i, g2 == g].mean() for g in set(g2) if g != propio}
vecino = min(b_por_grupo, key=b_por_grupo.get)
b = b_por_grupo[vecino]
sil = (b - a) / max(a, b)

fig, ax = plt.subplots(figsize=(11, 5.6))
for grupo, color in [(propio, TEAL), (vecino, AMBAR)]:
    for j in np.where(g2 == grupo)[0]:
        ax.plot([P2[i, 0], P2[j, 0]], [P2[i, 1], P2[j, 1]], color=color, alpha=0.22, lw=0.9, zorder=1)
colores_grupo = {propio: TEAL, vecino: AMBAR}
for g in sorted(set(g2)):
    ax.scatter(P2[g2 == g, 0], P2[g2 == g, 1], s=42, color=colores_grupo.get(g, GRIS_CLARO), linewidths=0, zorder=2)
ax.scatter([P2[i, 0]], [P2[i, 1]], s=280, color=MAGENTA, edgecolor="white", linewidth=2, zorder=4)
ax.annotate("este cliente", xy=P2[i], xytext=(14, -30), textcoords="offset points", ha="left",
            fontsize=11, color=MAGENTA, fontweight="semibold")
ax.text(0.02, 0.97, f"a = {a:.2f}   distancia promedio a su grupo\nb = {b:.2f}   distancia promedio al grupo vecino\n"
        f"silueta = (b - a) / máx(a, b) = {sil:.2f}", transform=ax.transAxes, va="top", fontfamily=LETRA_MONO,
        fontsize=10.5, color=TINTA, bbox=dict(fc="white", ec=LINEA, boxstyle="round,pad=0.6"))
ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
fig.tight_layout()
encabezado(fig, "Métricas · ejemplo", "La silueta compara dos distancias: a tu grupo (a) y al grupo vecino (b)",
           "Líneas azules: distancias a los de su grupo. Líneas ámbar: a los del grupo vecino más cercano. Si b es mucho mayor que a, el cliente está bien ubicado.",
           fuente="Datos simulados para el ejemplo")
plt.show()
```

Segundo: ¿la ventaja de un modelo depende de haber escogido 4 grupos? Calculamos las tres métricas de forma para los dos modelos, de 2 a 8 grupos.

<div style="border-left:3px solid #8E5BD9;background:#F0E7FA;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#6B3FB0;">EJECUTA</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">Esta celda entrena 14 modelos.
```python
filas = []
for k in range(2, 9):
    for nombre, modelo in [("K-means", KMeans(n_clusters=k, n_init=20, random_state=42)),
                           ("Jerárquico", AgglomerativeClustering(n_clusters=k, linkage="ward"))]:
        l = modelo.fit_predict(Z)
        filas.append({"k": k, "modelo": nombre, "Silueta": silhouette_score(Z, l),
                      "Calinski-Harabasz": calinski_harabasz_score(Z, l), "Davies-Bouldin": davies_bouldin_score(Z, l)})
por_k = pd.DataFrame(filas)

fig, ejes = plt.subplots(1, 3, figsize=(14, 4.8))
ganadas, total = 0, 0
for ax, m in zip(ejes, ["Silueta", "Calinski-Harabasz", "Davies-Bouldin"]):
    ancho = por_k.pivot(index="k", columns="modelo", values=m)
    ax.axvspan(K_ELEGIDO - 0.3, K_ELEGIDO + 0.3, color=MAGENTA_T, zorder=0)
    for nombre in ["K-means", "Jerárquico"]:
        ax.plot(ancho.index, ancho[nombre], marker="o", color=COLOR_MODELO[nombre], linewidth=2, label=nombre)
    gana_km = ancho["K-means"] < ancho["Jerárquico"] if m == "Davies-Bouldin" else ancho["K-means"] > ancho["Jerárquico"]
    ganadas += int(gana_km.sum()); total += len(gana_km)
    ax.set_title(m + ("  (más bajo es mejor)" if m == "Davies-Bouldin" else "  (más alto es mejor)"))
    ax.set_xlabel("Número de grupos (k)")
    if m == "Calinski-Harabasz":
        ax.yaxis.set_major_formatter(mpl.ticker.FuncFormatter(lambda v, _: miles(v)))
ejes[0].legend(loc="upper right")
fig.tight_layout(w_pad=2)
encabezado(fig, "Métricas · Fig. 5", f"K-means gana {ganadas} de {total} comparaciones entre 2 y 8 grupos",
           "La franja rosada marca k = 4, el número de grupos que escogimos. La ventaja de K-means no depende de esa elección, pero tampoco es total.")
plt.show()
```

## Los dos modelos finales, cara a cara

Ahora sí, las métricas de los dos modelos con k = 4, más la estabilidad.

<div style="border-left:3px solid #8E5BD9;background:#F0E7FA;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#6B3FB0;">EJECUTA</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">La celda de abajo entrena 40 modelos más (20 por cada método). Puede tardar unos 15 segundos.</div></div>

```python
def estabilidad(fabricar, etiquetas_base, repeticiones=20):
    """ARI entre el modelo original y modelos reentrenados con el 80 % de los clientes."""
    azar = np.random.default_rng(0)
    resultados = []
    for r in range(repeticiones):
        idx = azar.choice(len(Z), size=int(0.8 * len(Z)), replace=False)
        nuevas = fabricar(r).fit_predict(Z[idx])
        resultados.append(adjusted_rand_score(etiquetas_base[idx], nuevas))
    return np.array(resultados)

estab = {
    "K-means": estabilidad(lambda r: KMeans(n_clusters=K_ELEGIDO, n_init=10, random_state=r), kmeans.labels_),
    "Jerárquico": estabilidad(lambda r: AgglomerativeClustering(n_clusters=K_ELEGIDO, linkage="ward"), jerarquico.labels_),
}
etiquetas_modelo = {"K-means": kmeans.labels_, "Jerárquico": jerarquico.labels_}
metricas = pd.DataFrame({
    nombre: {"Silueta": silhouette_score(Z, l),
             "Calinski-Harabasz": calinski_harabasz_score(Z, l),
             "Davies-Bouldin": davies_bouldin_score(Z, l),
             "Estabilidad (ARI)": estab[nombre].mean()}
    for nombre, l in etiquetas_modelo.items()})
MEJOR_ALTO = {"Silueta": True, "Calinski-Harabasz": True, "Davies-Bouldin": False, "Estabilidad (ARI)": True}
metricas["Gana"] = [("K-means" if (fila["K-means"] > fila["Jerárquico"]) == MEJOR_ALTO[m] else "Jerárquico")
                    for m, fila in metricas.iterrows()]
tabla(metricas, formato={"K-means": "{:,.3f}", "Jerárquico": "{:,.3f}"})
```

La tabla dice mucho, pero se lee mejor como un tablero:

## La silueta, cliente por cliente

El promedio de la silueta esconde detalles. Este gráfico muestra la silueta de **cada cliente**, ordenada dentro de su segmento. Una barra larga es un cliente bien ubicado; una barra corta o negativa es un cliente en la frontera.

```python
fig, ejes = plt.subplots(1, 2, figsize=(13, 5.8), sharex=True)
for ax, (nombre, columna) in zip(ejes, [("K-means", "seg_km"), ("Jerárquico", "seg_jer")]):
    valores = silhouette_samples(Z, etiquetas_modelo[nombre])
    base = 0
    for seg in ORDEN_SEGMENTOS:
        s = np.sort(valores[(df[columna] == seg).values])
        ax.fill_betweenx(np.arange(base, base + len(s)), 0, s, color=COLOR_SEGMENTO[seg], alpha=0.9, linewidth=0)
        ax.text(-0.02, base + len(s) / 2, f"{seg}\n{s.mean():.2f}", ha="right", va="center", fontsize=9, color=TINTA)
        base += len(s) + 25
    ax.axvline(valores.mean(), color=TINTA, linestyle="--", linewidth=1.2)
    ax.text(valores.mean(), base, f" promedio {valores.mean():.2f}", fontsize=9.5, color=TINTA, va="bottom")
    ax.set_title(nombre, color=COLOR_MODELO[nombre])
    ax.set_yticks([]); ax.set_xlabel("Silueta de cada cliente"); ax.grid(axis="y", visible=False)
    ax.spines["left"].set_visible(False)
ejes[0].set_xlim(-0.25, 0.75)
fig.tight_layout(w_pad=12)
sil_seg = pd.Series(silhouette_samples(Z, kmeans.labels_), index=df.index).groupby(df["seg_km"]).mean()
encabezado(fig, "Métricas · Fig. 8", f"{sil_seg.idxmax()} es el segmento más nítido; {sil_seg.idxmin()}, el más difuso",
           "Cada línea horizontal es un cliente. Las colas que cruzan a la izquierda del cero son clientes que quedarían igual de bien en el segmento vecino.")
plt.show()
```

## ¿Los dos modelos están de acuerdo?

Cruzamos los segmentos de un modelo con los del otro. Si coincidieran por completo, todos los clientes estarían en la diagonal.

```python
cruce = pd.crosstab(df["seg_km"], df["seg_jer"])
acuerdo = adjusted_rand_score(df["grupo_km"], df["grupo_jer"])
en_diagonal = np.trace(cruce.values) / cruce.values.sum()

fig, ax = plt.subplots(figsize=(10, 5.6))
sns.heatmap(cruce, annot=True, fmt="d", cmap=sns.light_palette(MAGENTA, as_cmap=True), cbar=False,
            linewidths=2, linecolor="white", annot_kws={"fontsize": 12, "fontweight": "semibold"}, ax=ax)
ax.set_xlabel("Segmento según el jerárquico"); ax.set_ylabel("Segmento según K-means")
ax.set_xticklabels([textwrap.fill(t.get_text(), 16) for t in ax.get_xticklabels()], rotation=0)
ax.set_yticklabels([textwrap.fill(t.get_text(), 18) for t in ax.get_yticklabels()], rotation=0)
fig.tight_layout()
encabezado(fig, "Métricas · Fig. 9", f"Los dos modelos ubican igual al {en_diagonal:.0%} de los clientes",
           f"ARI entre modelos: {acuerdo:.2f}. Los desacuerdos se concentran en las fronteras: clientes que K-means llama Premium y el jerárquico manda a Familias consolidadas.")
plt.show()
```

## ¿Cuánto vale focalizar? Las cuentas de una campaña

El lift dice quién responde mejor, pero el gerente pregunta otra cosa: **¿cuánto dinero gano o pierdo?** La propia base trae dos columnas para hacer la cuenta: `Z_CostContact` (cuesta 3 contactar a un cliente) y `Z_Revenue` (cada cliente que acepta deja 11). Simulamos cuatro maneras de lanzar la próxima campaña, suponiendo que cada segmento responde como respondió a la última.

```python
COSTO, INGRESO = df["Z_CostContact"].iloc[0], df["Z_Revenue"].iloc[0]
respuestas_totales = df["Response"].sum()
escenarios = {
    "Campaña masiva (todos)": df.index == df.index,
    "Solo Premium (K-means)": df["seg_km"] == "Premium",
    "Solo Premium (jerárquico)": df["seg_jer"] == "Premium",
    "Premium + Familias consolidadas (K-means)": df["seg_km"].isin(["Premium", "Familias consolidadas"]),
}
cuentas = pd.DataFrame({
    nombre: {"Contactos": m.sum(), "Respuestas": df.loc[m, "Response"].sum()} for nombre, m in escenarios.items()
}).T
cuentas["Tasa de respuesta"] = cuentas["Respuestas"] / cuentas["Contactos"]
cuentas["Costo"] = cuentas["Contactos"] * COSTO
cuentas["Ingreso"] = cuentas["Respuestas"] * INGRESO
cuentas["Resultado"] = cuentas["Ingreso"] - cuentas["Costo"]
cuentas["Contactos por respuesta"] = cuentas["Contactos"] / cuentas["Respuestas"]
cuentas["Captura de quienes responden"] = cuentas["Respuestas"] / respuestas_totales
print(f"Costo por contacto: {COSTO} · Ingreso por respuesta: {INGRESO} · "
      f"Punto de equilibrio: tasa de respuesta de {COSTO / INGRESO:.1%}")
tabla(cuentas, formato={"Contactos": lambda v: miles(v), "Respuestas": lambda v: miles(v), "Tasa de respuesta": "{:.1%}",
                        "Costo": lambda v: miles(v), "Ingreso": lambda v: miles(v), "Resultado": lambda v: miles(v),
                        "Contactos por respuesta": "{:.1f}", "Captura de quienes responden": "{:.0%}"})
```

## Cómo se opera una segmentación en la empresa

Una segmentación no es un informe que se entrega una vez: es un proceso que se repite.

| Paso | Cada cuánto | Qué se hace | Qué se vigila |
|---|---|---|---|
| Recalcular los segmentos | Cada mes o cada trimestre | Volver a correr el modelo con los datos nuevos (mismas variables, mismo escalador) | Que la estabilidad siga alta y que los centros no se muevan mucho |
| Asignar clientes nuevos | Cada día | KNN (o el centro más cercano) le pone segmento a cada cliente que llega | Qué tan seguro está el modelo (la columna «Seguridad» de la sección 10) |
| Seguir las migraciones | Cada mes | Contar cuántos clientes pasan de un segmento a otro | Familias consolidadas → Premium es una buena noticia; Premium → cazadoras, una alarma |
| Medir cada campaña | Por campaña | Tasa de respuesta, lift y resultado por segmento | Que la campaña gane plata en el segmento donde se lanzó |
| Revisar el modelo | Cada año | Revisar si las variables siguen describiendo el negocio | Si aparece un quinto grupo o uno se vacía, toca rediseñar |

Sin un modelo estable, esa cuenta no tiene sentido, porque los segmentos cambiarían por el modelo y no por los clientes.</div></div>

# Escoger un modelo y ponerlo a trabajar

## La decisión, con criterios de negocio

Un analista no elige el modelo que «gana más métricas», sino el que **mejor sirve a la decisión**. Esta es la matriz completa:

| Criterio | K-means | Jerárquico (Ward) | ¿Quién gana? |
|---|---|---|---|
| Calidad de los grupos (silueta, CH, DB) | Ligeramente mejor | Ligeramente peor | K-means, por poco |
| **Estabilidad** mes a mes | Muy alta | Media | **K-means, claro** |
| Asignar clientes nuevos | Directo: centro más cercano | No tiene forma nativa; hay que entrenar otro modelo | K-means |
| Escala a millones de clientes | Sí, es rápido | Se vuelve lento y pesado en memoria | K-means |
| Explicarlo al equipo de marketing | «Cuatro clientes típicos y quién se parece a cada uno» | «Un árbol que cortamos en cuatro» | Empate |
| Ver segmentos y subsegmentos anidados | No | **Sí, el dendrograma** | Jerárquico |
| No depender del azar | Necesita semilla y repeticiones | Siempre da lo mismo con los mismos datos | Jerárquico |
| Segmento Premium | Más amplio: llega al 42 % de quienes responden, casi empata en plata | Más pequeño: llega al 35 %, pero es el único que gana plata | Depende del objetivo (alcance o rentabilidad) |

<div style="border-left:3px solid #E8943A;background:#FCE9D3;border-radius:0 14px 14px 0;padding:20px 26px;margin:18px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><div style="display:flex;align-items:center;gap:10px;margin-bottom:8px;"><span style="flex:none;width:30px;height:30px;border-radius:10px;background:#FCE9D3;display:inline-flex;align-items:center;justify-content:center;"><span style="display:flex;flex-wrap:wrap;gap:3px;width:15px;"><span style="width:6px;height:6px;background:#B26212;border-radius:1.5px;display:block"></span><span style="width:6px;height:6px;background:#B26212;border-radius:1.5px;display:block"></span><span style="width:6px;height:6px;background:#B26212;border-radius:1.5px;display:block"></span><span style="width:6px;height:6px;background:#B26212;border-radius:1.5px;display:block"></span></span></span><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:9.5pt;letter-spacing:2.5px;color:#B26212;">SÍNTESIS</span></div><div style="font-size:13pt;line-height:1.6;color:#2A2933;"><b>Escogemos K-means.</b> Los dos modelos encuentran los mismos cuatro tipos de cliente, así que la historia no depende del algoritmo. Lo que inclina la balanza es que K-means produce segmentos <b>estables</b>, que se pueden volver a calcular cada mes sin que cambien de forma, y que <b>asigna clientes nuevos</b> de manera directa. Para una campaña puntual y cara, en cambio, vale la pena mirar el corte del jerárquico: su Premium más pequeño fue el único rentable en la simulación. Y el jerárquico sería la mejor opción si el objetivo fuera <i>explorar</i> la estructura (cuántos niveles de segmentos tiene sentido usar) o si el negocio necesitara macro y microsegmentos a la vez.</div></div>

## Usar el resultado: asignar clientes nuevos con KNN

Mañana se registran clientes nuevos. ¿A qué segmento pertenecen? Aquí entra por fin **KNN** (*k vecinos más cercanos*), ahora en su papel correcto: el de **clasificador supervisado**. Los segmentos que descubrió K-means se vuelven la «respuesta» con la que KNN aprende.

La lógica es la de un buen vendedor: *«dime a qué clientes te pareces y te diré qué tipo de cliente eres»*. Para un cliente nuevo, KNN busca los clientes conocidos más parecidos y le asigna el segmento que tenga la mayoría.

<div style="border-left:3px solid #A3195B;background:#FBE8F1;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#7A1244;">DATO</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">K-means también podría asignar al cliente nuevo directamente a su centro más cercano. Usamos KNN porque es el método que sirve con <b>cualquier</b> segmentación (incluida la jerárquica, que no tiene forma propia de hacerlo) y porque nos deja medir qué tan confiable es la asignación con una métrica conocida: el porcentaje de aciertos.</div></div>

Primero elegimos cuántos vecinos mirar. Probamos de 1 a 25 con **validación cruzada**: se parten los datos en 5 pedazos, se entrena con 4 y se evalúa con el quinto, rotando.

```python
Z_train, Z_test, y_train, y_test = train_test_split(Z, df["seg_km"].astype(str), test_size=0.25,
                                                    random_state=42, stratify=df["seg_km"].astype(str))
vecinos = list(range(1, 26, 2))
aciertos = [cross_val_score(KNeighborsClassifier(n_neighbors=k), Z_train, y_train, cv=5).mean() for k in vecinos]
K_VECINOS = vecinos[int(np.argmax(aciertos))]

fig, ax = plt.subplots(figsize=(11, 4.6))
ax.plot(vecinos, aciertos, color=GRIS, marker="o", linewidth=2)
ax.scatter([K_VECINOS], [max(aciertos)], s=160, color=MAGENTA, zorder=3)
ax.annotate(f"{K_VECINOS} vecinos · {max(aciertos):.1%} de aciertos", xy=(K_VECINOS, max(aciertos)),
            xytext=(12, -24), textcoords="offset points", fontsize=10.5, color=MAGENTA, fontweight="semibold")
ax.set_xlabel("Número de vecinos (k de KNN)"); ax.set_ylabel("Aciertos en validación cruzada")
ax.yaxis.set_major_formatter(mpl.ticker.PercentFormatter(1, decimals=0))
ax.set_ylim(0.90, 1.0)
fig.tight_layout()
encabezado(fig, "Uso · Fig. 16", f"Con {K_VECINOS} vecinos, KNN reconoce el segmento de casi todos los clientes",
           "Pocos vecinos hacen al modelo nervioso (se deja llevar por un solo cliente raro); demasiados lo vuelven perezoso (todo se parece al promedio).")
plt.show()
```

```python
knn = KNeighborsClassifier(n_neighbors=K_VECINOS).fit(Z_train, y_train)
acierto_test = (knn.predict(Z_test) == y_test).mean()
print(f"Aciertos en clientes que el modelo nunca vio: {acierto_test:.1%}")
```

## Glosario

| Término | En una frase |
|---|---|
| **Aprendizaje no supervisado** | El modelo busca estructura en los datos sin que nadie le diga la respuesta |
| **Clustering / agrupamiento** | Juntar casos parecidos y separar los distintos |
| **K-means** | Agrupa alrededor de k centros que se mueven hasta estabilizarse |
| **Clustering jerárquico** | Fusiona casos de dos en dos hasta formar un árbol que luego se corta |
| **Ward** | Criterio de fusión que junta los grupos que menos aumentan la dispersión interna |
| **Dendrograma** | El dibujo del árbol del clustering jerárquico |
| **Escalar / estandarizar** | Poner todas las variables en la misma escala para que pesen igual |
| **Inercia** | Suma de distancias de los clientes a su centro (método del codo) |
| **Silueta** | Qué tan bien ubicado está cada caso en su grupo (−1 a 1, más es mejor) |
| **Calinski-Harabasz** | Separación entre grupos frente a dispersión interna (más es mejor) |
| **Davies-Bouldin** | Qué tanto se parece cada grupo a su vecino más cercano (menos es mejor) |
| **ARI** | Coincidencia entre dos agrupaciones (1 = idénticas, 0 = como al azar) |
| **PCA** | Resume muchas variables en pocas para poder dibujarlas |
| **KNN** | Clasifica un caso nuevo por mayoría de sus k vecinos más parecidos (supervisado) |
| **Lift** | Cuántas veces mejor responde un segmento frente al promedio |
| **Ticket promedio** | Gasto total ÷ número de compras: cuánto deja un cliente cada vez que compra |
| **Punto de equilibrio** | La tasa de respuesta a partir de la cual una campaña deja de perder plata |
| **Centroide** | El cliente típico de un grupo en K-means: el promedio de sus clientes |
| **Enlace (linkage)** | La regla del jerárquico para medir la distancia entre dos grupos |
