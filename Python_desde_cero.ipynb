# 1 · ¿Qué es Python?

Python es un **idioma para darle instrucciones a un computador**. Se parece más al inglés que a las matemáticas:
se lee de arriba hacia abajo, una línea a la vez, y cada línea hace **una cosa**.

Es el idioma más usado en ciencia de datos y en IA por tres razones que nos importan:

1. **Se lee casi como prosa**, así que puedes revisar lo que la IA te genera.
2. Tiene **librerías** para todo (ya veremos qué es eso).
3. Es lo que hablan las herramientas de ML: si pides un modelo, te lo van a entregar en Python.

La instrucción más básica es `print(...)`: «muéstrame esto en pantalla». Ejecuta la celda de abajo con **Shift + Enter**.

```python
print("Hola, Ciclo 4 👋")
```

Lo que está entre comillas es un **texto**. Lo que está fuera de las comillas es una **instrucción**.
Python también sabe de aritmética, y todo lo que empiece con `#` es un **comentario**: una nota para humanos que Python ignora.

```python
# Esto es un comentario. Python no lo lee, es para dejar notas.
print(2 + 2)
print(1500000 * 1.19)      # el IVA de un millón y medio
print("Python lee de arriba hacia abajo, una línea a la vez.")
```

# 2 · Variables: ponerle nombre a las cosas

Una **variable** es una caja con etiqueta. Guardas algo adentro y luego lo llamas por su nombre.
El signo `=` no significa «igual a», significa **«guarda esto aquí»**.

```python
campana = "Regreso a clases"
inversion = 14791921
roas = 4.23
activa = True

print(campana)
print(inversion)
print(roas)
print(activa)
```

## Los tipos de datos básicos

Cada caja tiene un **tipo**, y el tipo decide qué puedes hacer con ella: a un texto no se le puede sacar promedio,
y a un número no se le puede poner mayúsculas. **La mitad de los errores en datos son errores de tipo**:
un número que llegó como texto («1.500.000»), una fecha que es texto, un «Sí/No» que debería ser verdadero/falso.

| Tipo | En Python | Ejemplo | En Excel sería… |
|---|---|---|---|
| Texto | `str` (*string*) | `"Instagram"` | una celda con letras |
| Entero | `int` | `14791921` | un número sin decimales |
| Decimal | `float` | `4.23` | un número con decimales |
| Lógico | `bool` | `True` / `False` | una casilla de verificación |
| Vacío | `None` | `None` | una celda en blanco |

La función `type(...)` te dice de qué tipo es algo. Úsala siempre que un modelo se queje de tus datos.

```python
print(type(campana))
print(type(inversion))
print(type(roas))
print(type(activa))
```

### Combinar texto y números: las *f-strings*

Para armar una frase con variables adentro se pone una `f` antes de las comillas y las variables entre llaves `{ }`.
Es la forma en que la IA casi siempre te va a imprimir resultados.

```python
ingresos = inversion * roas
print(f"La campaña «{campana}» invirtió ${inversion:,.0f} y generó ${ingresos:,.0f} con un ROAS de {roas}.")
```

## Listas: varias cosas en orden

Una **lista** va entre corchetes `[ ]` y guarda varias cosas en orden. Se cuenta **desde cero**: el primer elemento es el `0`.
Esto es raro al principio y es la causa número uno de confusiones; la IA lo hace bien, tú solo tienes que saber que existe.

```python
canales = ["Instagram", "TikTok", "Google Ads", "Email", "Punto de venta"]

print(canales[0])        # el primero
print(canales[-1])       # el último
print(len(canales))      # cuántos hay

canales.append("WhatsApp")   # agregar uno al final
print(canales)
```

## Diccionarios: una ficha con campos

Un **diccionario** va entre llaves `{ }` y guarda pares **nombre: valor**. Es como la ficha de un cliente o de una campaña.
Cuando la IA te hable de «JSON», es esto mismo.

```python
ficha = {
    "campana": "Día de la Madre",
    "canal": "Instagram",
    "inversion": 18500000,
    "roas": 5.1,
    "regiones": ["Bogotá", "Medellín"],
}

print(ficha["canal"])
print(ficha["regiones"][1])
ficha["satisfaccion"] = 4.6          # agregar un campo nuevo
print(ficha)
```

# 3 · Tres ideas que verás en todo código: decidir, repetir, empaquetar

No necesitas escribirlas de memoria. Necesitas **reconocerlas** cuando aparezcan en lo que la IA te entregue.

## Decidir: `if` / `else`

«Si pasa esto, haz aquello; si no, haz lo otro». Fíjate en los **dos puntos** `:` y en la **sangría** (los espacios al inicio):
en Python, lo que está sangrado «pertenece» a la línea de arriba.

```python
roas = 4.23

if roas >= 4:
    print("Campaña rentable: escalar presupuesto")
elif roas >= 2:
    print("Campaña aceptable: optimizar creativos")
else:
    print("Campaña en pérdida: pausar")
```

## Repetir: `for`

«Para cada cosa de esta lista, haz esto». Así se recorren canales, meses, clientes… sin copiar y pegar.

```python
for canal in canales:
    print(f"Revisando resultados de {canal}...")
```

## Empaquetar: `def` (funciones)

Una **función** es una receta con nombre: le das ingredientes (entre paréntesis) y te devuelve un resultado con `return`.
`print()`, `len()` y `type()` son funciones que Python ya trae. Cuando la IA te genere código, casi siempre lo organiza en funciones.

```python
def clasificar_roas(roas):
    if roas >= 4:
        return "escalar"
    elif roas >= 2:
        return "optimizar"
    else:
        return "pausar"

print(clasificar_roas(5.1))
print(clasificar_roas(1.4))

# Y combinando las tres ideas:
resultados = {"Instagram": 5.1, "TikTok": 3.2, "Email": 1.4}
for canal, valor in resultados.items():
    print(f"{canal}: ROAS {valor} → {clasificar_roas(valor)}")
```

# 5 · NumPy: números en bloque

Una lista de Python sabe guardar números, pero no sabe hacer matemáticas con todos a la vez.
NumPy convierte la lista en un **arreglo** (`array`) y entonces una operación se aplica a **todos los elementos de un golpe**.
Esto se llama *vectorizar* y es la razón de que pandas y scikit-learn sean rápidos: por debajo, todos usan NumPy.

```python
ventas_semana = [120, 135, 98, 160, 142, 175, 130]          # una lista normal
ventas = np.array(ventas_semana)                             # ahora es un arreglo de NumPy

print("Promedio:", ventas.mean())
print("Máximo:", ventas.max(), "| Mínimo:", ventas.min())
print("Desviación estándar:", ventas.std().round(1))
print("Con 10 % de aumento:", ventas * 1.10)                 # una operación, los siete valores
print("Días por encima del promedio:", (ventas > ventas.mean()).sum())
```

NumPy también genera **datos al azar** con reglas: útil para simular escenarios («¿y si la conversión fuera 3 % ± 1 %?»).
La **semilla** (`seed`) hace que el azar sea repetible: todo el salón obtiene los mismos números.

```python
rng = np.random.default_rng(seed=42)
conversiones_simuladas = rng.normal(loc=3.0, scale=1.0, size=1000)   # 1000 escenarios, media 3 %, desviación 1

print("Promedio simulado:", conversiones_simuladas.mean().round(2), "%")
print("Peor 5 % de escenarios por debajo de:", np.percentile(conversiones_simuladas, 5).round(2), "%")
```

# 6 · pandas: tablas que obedecen

**pandas** es la librería con la que vas a pasar el 80 % del tiempo. Su objeto estrella es el **DataFrame**:
una tabla con filas y columnas, exactamente como una hoja de Excel, pero que se manipula con instrucciones en vez de clics.

## Filtrar, agrupar, calcular

Esto es lo que en Excel harías con filtros y tablas dinámicas. Fíjate en que cada instrucción se lee casi en voz alta.

```python
# Filtrar: solo las filas de Instagram en Bogotá
instagram_bogota = df[(df["canal"] == "Instagram") & (df["region"] == "Bogotá")]
print("Campañas de Instagram en Bogotá:", len(instagram_bogota))

# Columna calculada: ganancia en millones
df["ganancia_millones"] = (df["ingresos"] - df["inversion"]) / 1_000_000

# Agrupar: la tabla dinámica de ingresos por canal
ingresos_por_canal = df.groupby("canal")["ingresos"].sum().sort_values(ascending=False) / 1_000_000
print("\nIngresos por canal (millones de pesos):")
print(ingresos_por_canal.round(1))
```

```python
# Varias métricas a la vez, por región
df.groupby("region").agg(
    inversion_total=("inversion", "sum"),
    ingresos_total=("ingresos", "sum"),
    roas_promedio=("roas", "mean"),
    satisfaccion=("satisfaccion", "mean"),
).round(2).sort_values("ingresos_total", ascending=False)
```

# 7 · matplotlib: la anatomía de una gráfica

**matplotlib** es la librería base de gráficas en Python. Casi todo lo que se dibuja en Python (incluido seaborn) termina en matplotlib,
así que vale la pena conocer sus **cuatro piezas**, porque son las que vas a pedir cambiar en los prompts:

| Pieza | Qué es | Cómo se llama en código |
|---|---|---|
| **Figura** | El lienzo completo | `fig` |
| **Ejes** | El área donde van los datos (una figura puede tener varios) | `ax` |
| **Marcas** | Barras, líneas, puntos | `ax.bar(...)`, `ax.plot(...)`, `ax.scatter(...)` |
| **Textos** | Título, etiquetas de ejes, anotaciones | `ax.set_title(...)`, `ax.set_xlabel(...)`, `ax.annotate(...)` |

La frase `fig, ax = plt.subplots()` es «dame un lienzo con un área para dibujar». Todo lo demás se le dice a `ax`.

```python
fig, ax = plt.subplots()

ax.bar(ingresos_por_canal.index, ingresos_por_canal.values, color=PALETA[0], width=0.6)

ax.set_title("Punto de venta, Instagram y TikTok concentran tres cuartos de los ingresos")
ax.set_ylabel("Ingresos 2025 (millones de pesos)")
ax.set_xlabel("")

# Etiquetas directas encima de cada barra: el lector no tiene que leer el eje
for i, valor in enumerate(ingresos_por_canal.values):
    ax.text(i, valor + 60, f"{valor:,.0f}", ha="center", va="bottom", fontsize=10)

plt.show()
```

Una **línea de tiempo** con dos piezas más: una línea de referencia y una anotación. Fíjate en que primero se **prepara la tabla**
con pandas (agrupar por mes) y después se **dibuja** con matplotlib: ese es siempre el orden.

```python
ingresos_mes = df.groupby("mes")["ingresos"].sum() / 1_000_000
meses = df.drop_duplicates("mes").set_index("mes")["nombre_mes"]       # Ene, Feb, ... para el eje

fig, ax = plt.subplots()
ax.plot(meses.values, ingresos_mes.values, color=PALETA[0], linewidth=2, marker="o", markersize=6)
ax.axhline(ingresos_mes.mean(), color="gray", linestyle="--", linewidth=1)
ax.text(0, ingresos_mes.mean() + 20, f"Promedio mensual: {ingresos_mes.mean():,.0f}", color="gray", fontsize=10)

pico = ingresos_mes.idxmax()
ax.annotate("Black Friday + Navidad", xy=(list(ingresos_mes.index).index(pico), ingresos_mes.max()),
            xytext=(6, ingresos_mes.max() * 0.9), fontsize=10,
            arrowprops=dict(arrowstyle="->", color="gray"))

ax.set_title("Los ingresos se disparan en el último trimestre")
ax.set_ylabel("Ingresos (millones de pesos)")
ax.set_ylim(0)
plt.show()
```

# 8 · seaborn: gráficas estadísticas con una línea

**seaborn** vive encima de matplotlib y resuelve lo que en matplotlib tomaría veinte líneas: comparar grupos, ver distribuciones,
cruzar variables con color. Su truco es que **entiende el DataFrame**: le dices `data=df`, qué columna va en `x`, cuál en `y`
y cuál define el color (`hue`), y él agrupa, promedia y dibuja.

Como está encima de matplotlib, todo lo de la sección anterior (título, etiquetas, anotaciones) sigue funcionando igual.

```python
fig, ax = plt.subplots()
sns.barplot(data=df, x="canal", y="roas", hue="canal", legend=False, errorbar=None, width=0.6, ax=ax)
ax.set_title("El Email tiene el ROAS promedio más alto, aunque mueve menos plata")
ax.set_ylabel("ROAS promedio")
ax.set_xlabel("")
plt.show()
```

### Ver la distribución, no solo el promedio

# 9 · scikit-learn: Machine Learning en tres pasos

**scikit-learn** es la librería de Machine Learning «clásico»: predecir un número, clasificar en categorías, agrupar clientes parecidos.
Todo modelo de scikit-learn, sin importar cuál, se usa con **la misma receta de tres pasos**. Si entiendes esta receta,
entiendes el 90 % del código de ML que la IA te va a generar.

<table style="font-size:15px">
<tr><th>Paso</th><th>Qué se hace</th><th>En código</th></tr>
<tr><td><b>1. Elegir el modelo</b></td><td>Decidir qué tipo de patrón buscamos</td><td><code>modelo = LinearRegression()</code></td></tr>
<tr><td><b>2. Entrenar</b></td><td>Mostrarle datos para que aprenda el patrón</td><td><code>modelo.fit(X, y)</code></td></tr>
<tr><td><b>3. Predecir</b></td><td>Usarlo con datos nuevos</td><td><code>modelo.predict(X_nuevo)</code></td></tr>
</table>

Dos nombres que vas a ver siempre: **`X`** (mayúscula) son las columnas con las que predecimos, **`y`** (minúscula) es lo que queremos predecir.

## 9.1 · Predecir un número: regresión lineal

Pregunta de negocio: *«si invierto 20 millones, ¿cuántos ingresos espero?»*. El modelo más simple traza la mejor línea recta posible entre inversión e ingresos.

Antes de entrenar, se **separa** la tabla: una parte para que el modelo aprenda (*train*) y otra que el modelo nunca vio, para evaluarlo honestamente (*test*).
Es como estudiar con unos ejercicios y presentar el examen con otros.

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

X = df[["inversion"]]          # con qué predecimos (una tabla, aunque sea de una columna)
y = df["ingresos"]             # qué queremos predecir

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

modelo = LinearRegression()            # 1. elegir
modelo.fit(X_train, y_train)           # 2. entrenar

predicciones = modelo.predict(X_test)  # 3. predecir sobre lo que nunca vio
print(f"Cada peso invertido se traduce en ${modelo.coef_[0]:.2f} de ingresos (la pendiente de la recta).")
print(f"R² en datos nuevos: {r2_score(y_test, predicciones):.2f}  (1.0 sería perfecto; 0 sería adivinar)")

inversion_nueva = pd.DataFrame({"inversion": [20_000_000]})
print(f"Si invierto $20.000.000, el modelo espera ingresos de ${modelo.predict(inversion_nueva)[0]:,.0f}")
```

```python
fig, ax = plt.subplots(figsize=(9, 5.2))
ax.scatter(X_test["inversion"] / 1e6, y_test / 1e6, color=PALETA[0], s=45, alpha=0.75, label="Campañas reales (test)")
orden = X_test["inversion"].sort_values()
ax.plot(orden / 1e6, modelo.predict(orden.to_frame()) / 1e6, color=PALETA[1], linewidth=2.5, label="Lo que el modelo predice")
ax.set_title("La recta captura la tendencia; la dispersión es lo que la inversión sola no explica")
ax.set_xlabel("Inversión (millones de pesos)")
ax.set_ylabel("Ingresos (millones de pesos)")
ax.legend(frameon=False)
plt.show()
```

## 9.2 · Encontrar grupos: clustering con K-Means

La Semana 4 fue de **clustering**. Acá lo ves en su forma mínima: le damos al modelo varias características de cada campaña
y le pedimos que las agrupe en `k` grupos de campañas **parecidas entre sí**, sin decirle cuáles son. Eso es aprendizaje **no supervisado**:
no hay `y`, no hay respuesta correcta; el modelo descubre la estructura.

Un detalle que la IA siempre incluye y ahora vas a entender: **escalar** los datos (`StandardScaler`). La inversión está en millones y la satisfacción
va de 1 a 5; sin escalar, el modelo solo «vería» la inversión. Escalar pone todas las columnas en la misma vara.

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

caracteristicas = ["inversion", "roas", "tasa_conversion", "satisfaccion"]
X_grupos = StandardScaler().fit_transform(df[caracteristicas])      # misma vara para todas las columnas

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)           # 1. elegir (3 grupos)
df["segmento"] = kmeans.fit_predict(X_grupos)                        # 2 + 3. entrenar y asignar grupo a cada fila

# ¿Qué caracteriza a cada grupo? El promedio de cada variable por segmento
perfil = df.groupby("segmento")[caracteristicas].mean().round(2)
perfil["n_campanas"] = df["segmento"].value_counts().sort_index()
perfil["canal_mas_comun"] = df.groupby("segmento")["canal"].agg(lambda s: s.value_counts().index[0])
perfil
```

```python
fig, ax = plt.subplots(figsize=(9, 5.2))
sns.scatterplot(data=df, x=df["inversion"] / 1e6, y="roas", hue="segmento", palette=PALETA[:3], s=60, alpha=0.85,
                edgecolor="white", linewidth=0.8, ax=ax)
ax.set_title("Sin saber el canal, el modelo redescubrió tres familias: Email, punto de venta y pauta digital")
ax.set_xlabel("Inversión (millones de pesos)")
ax.set_ylabel("ROAS")
ax.legend(title="Segmento", frameon=False)
plt.show()
```

# 10 · Promptear Machine Learning desde Colab

Ya sabes lo suficiente para pedirle a una IA un análisis completo **y entender lo que te devuelve**. Lo que diferencia un prompt
que funciona de uno que no es la **estructura**, no la elocuencia. Un buen prompt de análisis de datos tiene cinco partes:

| Parte | Qué pones | Ejemplo |
|---|---|---|
| **1. Contexto** | Quién eres y qué decisión vas a tomar | «Soy gerente de marketing y debo decidir dónde redistribuir presupuesto» |
| **2. Datos** | Nombre del DataFrame, columnas y tipos | «Tengo un DataFrame `df` con las columnas…» (la celda de abajo te lo arma) |
| **3. Tarea** | Qué quieres exactamente, en términos de ML si los conoces | «Entrena un modelo que prediga `ingresos` a partir de inversión, canal y región» |
| **4. Restricciones** | Librerías, estilo, qué no hacer | «Usa scikit-learn y seaborn. Usa la paleta `PALETA`. No inventes columnas.» |
| **5. Formato** | Cómo quieres la respuesta | «Código en una sola celda, comentado línea por línea, y al final una interpretación de 3 frases para un gerente» |

**Dónde pegar el prompt:** si tu Colab muestra el botón **Generar** o el ícono de Gemini en una celda de código, pégalo ahí y el código aparece en la celda.
Si no, pégalo en Gemini, ChatGPT o Claude, copia el código que te devuelva y pégalo en una celda **+ Código** de este cuaderno.

Esta celda arma la **parte 2** por ti: un resumen de tus datos listo para copiar en cualquier prompt. Úsala con cualquier tabla, siempre.

```python
def describir_para_prompt(df, nombre="df"):
    """Arma la descripción de un DataFrame para pegarla en un prompt."""
    lineas = [f"Tengo un DataFrame de pandas llamado `{nombre}` con {df.shape[0]} filas y {df.shape[1]} columnas:"]
    for col in df.columns:
        if df[col].dtype == "object" or str(df[col].dtype) == "str":
            ejemplos = ", ".join(map(str, df[col].unique()[:5]))
            lineas.append(f"- `{col}` (texto/categoría): valores como {ejemplos}")
        else:
            lineas.append(f"- `{col}` (numérica): entre {df[col].min():,.2f} y {df[col].max():,.2f}, promedio {df[col].mean():,.2f}")
    return "\n".join(lineas)

print(describir_para_prompt(df.drop(columns=["ganancia_millones", "segmento"])))
```
