# El precio de la casilla más cara

En el juego del umbral, la casilla que más dolía era **la fuga que no vi**: un cliente que se fue sin que lo llamáramos. Le pusimos un precio simple, 12 meses de su cargo. Pero en muchos negocios no hay un «cargo mensual» fijo: el cliente gasta lo que quiere, cuando quiere. ¿Cuánto vale entonces el que estoy a punto de perder? ¿Y cuánto puedo gastar en retenerlo?

Volvemos a la cadena de retail de la Semana 5 (vinos, carnes, pescados, frutas, dulces, oro; tienda, web y catálogo). Ya sabemos **qué tipos de clientes** tiene. Hoy la pregunta es otra:

> *«Si me llega un cliente nuevo, con su ingreso, su familia y su antigüedad, ¿cuánto va a gastar con nosotros? Y a los que ya tengo, ¿cuánto vale cada uno?»*

# Predecir una cifra, no un sí o no

El miércoles el modelo respondía **sí o no** (se va o se queda): eso es **clasificar**. Hoy responde **un número** (cuánto va a gastar): eso es **predecir una cifra** o, con el nombre técnico, hacer una **regresión**. Las dos son supervisadas: hay historia con la respuesta conocida, se separa en entrenar y probar, y se mide en lo que el modelo no vio.

Lo que cambia es **cómo se mide el error**. Ya no hay casillas de acierto o falsa alarma: hay una distancia entre lo que el modelo dijo y lo que pasó. Si dijo 800 USD y el cliente gastó 650, se equivocó por 150. El promedio de esas distancias, en dólares, es el **error promedio**.

## El mapa de las tres preguntas

| Pregunta de negocio | Tarea | Hay etiqueta | Modelos |
|---|---|---|---|
| ¿Quiénes se parecen? | Agrupar | No | K-means, Ward (Semana 5) |
| ¿Se va o se queda? | Clasificar | Sí: se fue o no | Regresión logística, árbol (miércoles) |
| **¿Cuánto vale?** | **Predecir una cifra** | **Sí: cuánto gastó** | **Regresión lineal, árbol (hoy)** |

La limpieza es la misma de la Semana 5: calculamos la edad (el último registro es de 2014), quitamos a quien no tiene ingreso, a un ingreso imposible y a edades imposibles. Sumamos el gasto de las seis categorías y contamos los hijos en casa.

```python
URL = "https://raw.githubusercontent.com/YourFavouriteDataSuperstar/Data-storytelling/main/clientes_marketing_campaign.csv"
df = pd.read_csv(URL, sep="\t")
df["Edad"] = 2014 - df["Year_Birth"]
df = df.dropna(subset=["Income"])
df = df[(df["Edad"] <= 90) & (df["Income"] < 200_000)].copy()
productos = ["MntWines", "MntFruits", "MntMeatProducts", "MntFishProducts", "MntSweetProducts", "MntGoldProds"]
df["Ingreso"] = df["Income"]
df["Gasto"] = df[productos].sum(axis=1)
df["Hijos"] = df["Kidhome"] + df["Teenhome"]
fecha = pd.to_datetime(df["Dt_Customer"], format="%d-%m-%Y")
df["Antiguedad"] = ((pd.Timestamp("2014-12-31") - fecha).dt.days / 30.44).round().astype(int)
df = df.reset_index(drop=True)
print(f"{len(df):,} clientes. Gasto promedio en dos años: {df.Gasto.mean():,.0f} USD; mediano: {df.Gasto.median():,.0f} USD.".replace(",", "."))
tabla(df[["Ingreso", "Hijos", "Antiguedad", "Edad", "Gasto"]].head(6).rename(columns={"Antiguedad": "Antigüedad (meses)", "Ingreso": "Ingreso anual"}),
      formato={"Ingreso anual": "{:,.0f} USD", "Gasto": "{:,.0f} USD"})
```

## Separar antes de tocar nada

```python
from sklearn.model_selection import train_test_split
PISTAS = ["Ingreso", "Hijos", "Antiguedad", "Edad"]
X_entrenar, X_probar, y_entrenar, y_probar = train_test_split(df[PISTAS], df["Gasto"], test_size=0.2, random_state=42)
print(f"Entrenar: {len(X_entrenar):,} clientes · Probar: {len(X_probar):,} clientes".replace(",", "."))
```

# Lo que dibujaron a mano, hecho por el computador

En clase dibujaron a mano una recta sobre la nube de puntos de **ingreso contra gasto**. El computador hace lo mismo, pero con una regla: elige la recta que deja las distancias verticales a los puntos **lo más pequeñas posible**. Eso es la **regresión lineal** con una pista.

La recta magenta es la del computador. Las líneas finas son el error de cada cliente: la distancia entre lo que la recta dijo y lo que gastó de verdad. El error promedio es el promedio de esas distancias.</div></div>

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

recta = LinearRegression().fit(X_entrenar[["Ingreso"]], y_entrenar)
pred_recta = recta.predict(X_probar[["Ingreso"]])
error_recta = mean_absolute_error(y_probar, pred_recta)
error_promedio_a_secas = mean_absolute_error(y_probar, np.full(len(y_probar), y_entrenar.mean()))

prueba = X_probar.copy(); prueba["Gasto"] = y_probar.values; prueba["Recta"] = pred_recta
muestra = prueba[prueba.Ingreso.between(10_000, 110_000)].sample(40, random_state=3)

fig, ax = plt.subplots(figsize=(10.5, 5.2))
ax.scatter(prueba.Ingreso, prueba.Gasto, s=9, color=GRIS, alpha=0.35)
malla = np.linspace(prueba.Ingreso.min(), prueba.Ingreso.max(), 50)
ax.plot(malla, recta.predict(pd.DataFrame({"Ingreso": malla})), color=MAGENTA, lw=3, label="La recta del computador")
for _, r in muestra.iterrows():
    ax.plot([r.Ingreso, r.Ingreso], [r.Gasto, r.Recta], color=AMBAR, lw=1, alpha=0.9)
ax.scatter(muestra.Ingreso, muestra.Gasto, s=22, color=TINTA, zorder=3, label="40 clientes con su error dibujado")
ax.axhline(0, color=GRIS_CLARO, lw=1)
ax.set_xlabel("Ingreso anual del hogar (USD)"); ax.set_ylabel("Gasto en dos años (USD)")
ax.xaxis.set_major_formatter(mpl.ticker.FuncFormatter(lambda v, _: f"{v/1000:.0f} mil"))
ax.legend(loc="upper left")
encabezado(fig, "Figura 1 · La recta y sus errores",
           f"La recta se equivoca en promedio por {error_recta:,.0f} USD; adivinar «el promedio» se equivoca por {error_promedio_a_secas:,.0f}".replace(",", "."),
           "Clientes del 20 % de prueba; las líneas ámbar son la distancia entre la recta y el gasto real")
plt.show()
print("Por cada 10.000 USD más de ingreso, la recta suma", miles(recta.coef_[0] * 10_000), "USD de gasto.")
print("Lo que la recta dice para un ingreso de 20.000 USD:", miles(recta.predict(pd.DataFrame({'Ingreso': [20_000]}))[0]), "USD")
```

# Ingreso, hijos, antigüedad y edad

Una sola pista deja mucho afuera. Con cuatro, la recta se vuelve una **suma de pesos**, igual que la tarjeta de puntos del miércoles, pero ahora el resultado es un gasto en dólares en vez de una probabilidad.

<div style="border-left:3px solid #1B7E9C;background:#E1F0F5;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#14657D;">CÓMO SE LEE</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">Cada barra dice cuánto cambia el gasto predicho cuando esa pista sube una unidad razonable: 10.000 USD más de ingreso, un hijo más en casa, un año más de antigüedad, diez años más de edad.</div></div>

```python
recta4 = LinearRegression().fit(X_entrenar, y_entrenar)
pred_recta4 = recta4.predict(X_probar)
error_recta4 = mean_absolute_error(y_probar, pred_recta4)

unidades = {"Ingreso": (10_000, "+10.000 USD de ingreso"), "Hijos": (1, "un hijo más en casa"),
            "Antiguedad": (12, "un año más de antigüedad"), "Edad": (10, "diez años más de edad")}
efectos = pd.Series({nombre: recta4.coef_[i] * unidad for i, (p, (unidad, nombre)) in enumerate(unidades.items())}).sort_values()

fig, ax = plt.subplots(figsize=(10, 3.8))
ax.barh(efectos.index, efectos.values, color=[ROJO if v < 0 else TEAL for v in efectos.values])
ax.axvline(0, color=TINTA, lw=1)
for i, v in enumerate(efectos.values):
    ax.text(v + (8 if v > 0 else -8), i, f"{v:+,.0f} USD".replace(",", "."), va="center", ha="left" if v > 0 else "right", fontsize=10)
ax.set_xticks([]); ax.grid(False)
encabezado(fig, "Figura 2 · Qué pesa en el gasto",
           f"Con cuatro pistas el error promedio baja a {error_recta4:,.0f} USD; los hijos en casa restan, el ingreso y la antigüedad suman".replace(",", "."),
           "Cambio en el gasto predicho por unidad de cada pista; regresión lineal con el 80 % de entrenamiento")
plt.show()
```

# Escalones en vez de recta

El árbol de decisión también sirve para predecir una cifra. Pregunta («¿ingreso mayor a 60.000?», «¿hijos en casa?», «¿más de 20 meses de antigüedad?») y en cada hoja responde con **el gasto promedio de los clientes que llegaron ahí**. Dibujado sobre la nube, en vez de una recta sale una **escalera**.

<div style="border-left:3px solid #1B7E9C;background:#E1F0F5;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#14657D;">CÓMO SE LEE</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">Compara las dos formas sobre la misma nube: la recta sube siempre al mismo ritmo; los escalones se quedan planos donde la nube es plana y saltan donde la nube salta. Por eso aquí el árbol se equivoca menos.</div></div>

```python
from sklearn.tree import DecisionTreeRegressor, plot_tree

arbol1 = DecisionTreeRegressor(max_depth=3, min_samples_leaf=40, random_state=42).fit(X_entrenar[["Ingreso"]], y_entrenar)
arbol4 = DecisionTreeRegressor(max_depth=3, min_samples_leaf=40, random_state=42).fit(X_entrenar, y_entrenar)
error_arbol1 = mean_absolute_error(y_probar, arbol1.predict(X_probar[["Ingreso"]]))
error_arbol4 = mean_absolute_error(y_probar, arbol4.predict(X_probar))

fig, ax = plt.subplots(figsize=(10.5, 5.2))
ax.scatter(prueba.Ingreso, prueba.Gasto, s=9, color=GRIS, alpha=0.35)
ax.plot(malla, recta.predict(pd.DataFrame({"Ingreso": malla})), color=MAGENTA, lw=2.5, label=f"Recta (error {error_recta:,.0f} USD)".replace(",", "."))
ax.step(malla, arbol1.predict(pd.DataFrame({"Ingreso": malla})), where="post", color=TEAL, lw=3, label=f"Árbol (error {error_arbol1:,.0f} USD)".replace(",", "."))
ax.axhline(0, color=GRIS_CLARO, lw=1)
ax.set_xlabel("Ingreso anual del hogar (USD)"); ax.set_ylabel("Gasto en dos años (USD)")
ax.xaxis.set_major_formatter(mpl.ticker.FuncFormatter(lambda v, _: f"{v/1000:.0f} mil"))
ax.legend(loc="upper left")
encabezado(fig, "Figura 3 · Recta contra escalones",
           "Donde la nube es plana y luego salta, los escalones del árbol se equivocan menos que la recta",
           "Los dos modelos usan solo el ingreso; error promedio medido en el 20 % de prueba")
plt.show()
```

# Ponerle precio a la fuga y a la oferta

Ahora cerramos el círculo con el miércoles. El valor predicho de un cliente sirve para dos cosas:

1. **El precio de la fuga que no vi** ya no es un número fijo: es lo que ese cliente iba a gastar.
2. **El tamaño de la oferta** se decide con ese valor: a un cliente que vale 1.500 USD se le puede ofrecer más que a uno que vale 90.

<div style="border-left:3px solid #2E8B57;background:#E3F2E9;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#1F6B42;">EN PLATA</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">Regla sencilla para el plan de retención: la oferta no debería pasar de una fracción del valor predicho del cliente (por ejemplo, el 15 %). Si el modelo de churn dice 70 % de riesgo y el de valor dice 1.200 USD, la pérdida esperada es 840 USD y una oferta de 150 se paga sola.</div></div>

```python
clientes_nuevos = pd.DataFrame([
    {"Ingreso": 85_000, "Hijos": 0, "Antiguedad": 30, "Edad": 52},
    {"Ingreso": 45_000, "Hijos": 2, "Antiguedad": 10, "Edad": 38},
    {"Ingreso": 62_000, "Hijos": 1, "Antiguedad": 24, "Edad": 45},
])
clientes_nuevos["Valor según la recta"] = recta4.predict(clientes_nuevos[PISTAS])
clientes_nuevos["Valor según el árbol"] = arbol4.predict(clientes_nuevos[PISTAS])
clientes_nuevos["Oferta máxima (15 %)"] = 0.15 * clientes_nuevos["Valor según el árbol"]
tabla(clientes_nuevos.rename(columns={"Antiguedad": "Antigüedad"}),
      formato={"Ingreso": "{:,.0f} USD", "Valor según la recta": "{:,.0f} USD", "Valor según el árbol": "{:,.0f} USD", "Oferta máxima (15 %)": "{:,.0f} USD"})
```
