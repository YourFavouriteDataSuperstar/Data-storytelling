# Una telco que se entera tarde de que el cliente se fue

Una empresa de telecomunicaciones vende teléfono e internet a hogares. Cada mes, una parte de sus clientes **cancela**: se va a la competencia o simplemente deja el servicio. En el sector a eso le dicen **churn** (fuga). Cuando la empresa se entera, ya es tarde: el cliente ya se fue y traerlo de vuelta cuesta mucho más que haberlo retenido.

El equipo de retención tiene un presupuesto limitado para **llamar** a clientes y ofrecerles algo (dos meses gratis, por ejemplo). No puede llamar a todos. La pregunta es la misma del juego de clase:

> *«¿A quién llamo primero, y desde qué nivel de riesgo vale la pena llamar?»*

La empresa no tiene nombre; la historia de la telco la ponemos nosotros.</div></div>

# Aprender con la respuesta en la mano

La semana pasada, K-means y Ward **descubrían** grupos de clientes sin que nadie les dijera la respuesta: eso es aprendizaje **no supervisado**. Hoy es distinto: la telco **sí sabe** quién se fue. Esa respuesta conocida se llama **etiqueta**, y un modelo que aprende de datos con etiqueta es **supervisado**.

La idea cabe en cuatro pasos:

1. **Historia con etiqueta.** Miles de clientes del pasado y, de cada uno, lo que sabíamos de él y si se fue.
2. **Limpiar.** Celdas vacías, texto donde iba un número, categorías mal escritas.
3. **Separar.** Una parte para **entrenar** (el modelo la estudia) y otra para **probar** (el modelo nunca la ve).
4. **Medir en lo que no vio.** Si el modelo estudia con el examen, saca 100 y no aprendió nada. Lo único que vale es cómo le va con clientes que no conocía.

## Limpiar antes de modelar

Dos cosas típicas pasan en esta base y pasan en casi cualquier base real:

* `TotalCharges` (lo que el cliente ha pagado en total) viene como **texto**, y en 11 clientes recién llegados está **vacío**. Lo convertimos a número y sacamos esos 11: todavía no tienen historia.
* `SeniorCitizen` (adulto mayor) viene como 0 y 1 en vez de *No* y *Yes*. Lo escribimos como las demás columnas para que se lea igual.

Y creamos la etiqueta en números: `se_fue` vale 1 si el cliente se fue y 0 si se quedó.

```python
datos["TotalCharges"] = pd.to_numeric(datos["TotalCharges"], errors="coerce")
vacios = datos["TotalCharges"].isna().sum()
datos = datos.dropna(subset=["TotalCharges"]).copy()
datos["SeniorCitizen"] = datos["SeniorCitizen"].map({0: "No", 1: "Yes"})
datos["se_fue"] = (datos["Churn"] == "Yes").astype(int)
print(f"Se quitaron {vacios} clientes sin cobro total. Quedan {len(datos):,} clientes.".replace(",", "."))
print(f"Se fue el {datos.se_fue.mean():.1%}; se quedó el {1 - datos.se_fue.mean():.1%}.")
```

## Mirar antes de modelar

Antes de cualquier modelo conviene ver **de qué tamaño es el problema** y **dónde se concentra**. Tres preguntas rápidas: ¿cuántos se van?, ¿importa el tipo de contrato?, ¿importa la antigüedad?

# El 80 % para entrenar, el 20 % para probar

Separamos **antes** de entrenar. El 20 % de prueba se guarda bajo llave: el modelo no lo ve hasta el final, y todo lo que midamos (aciertos, falsas alarmas, plata) se mide ahí.

Dos detalles que importan:

* **Al azar, pero con la misma proporción.** En los dos pedazos se va más o menos uno de cada cuatro. Si no, el examen no se parecería a la vida real.
* **Con semilla.** Un número fijo (`random_state=42`) hace que el sorteo salga igual cada vez que corras el cuaderno. Sin eso, cada corrida daría cifras distintas y no podríamos hablar de las mismas.

<div style="border-left:3px solid #1B7E9C;background:#E1F0F5;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#14657D;">CÓMO SE LEE</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">Las pistas que usamos: antigüedad y cargo mensual (números) y contrato, internet, forma de pago, soporte técnico, seguridad en línea, factura sin papel, pareja, personas a cargo y adulto mayor (categorías). Dejamos fuera lo que no cambia una decisión (el género) y lo que es consecuencia de la antigüedad (el cobro total).</div></div>

```python
from sklearn.model_selection import train_test_split

NUMERICAS = ["tenure", "MonthlyCharges"]
CATEGORICAS = ["Contract", "InternetService", "PaymentMethod", "TechSupport", "OnlineSecurity",
               "PaperlessBilling", "Partner", "Dependents", "SeniorCitizen"]
X = datos[NUMERICAS + CATEGORICAS]
y = datos["se_fue"]
X_entrenar, X_probar, y_entrenar, y_probar = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
print(f"Entrenar: {len(X_entrenar):,} clientes (se va el {y_entrenar.mean():.1%})".replace(",", "."))
print(f"Probar:   {len(X_probar):,} clientes (se va el {y_probar.mean():.1%})".replace(",", "."))
```

# La regresión logística es una tarjeta de puntos

Piensa en la tarjeta de puntos de una aerolínea, pero al revés: cada dato del cliente **suma o resta puntos de riesgo**. Contrato mes a mes suma. Cinco años de antigüedad restan. Fibra óptica con cargo alto suma. Al final, la suma de puntos se convierte en una **probabilidad entre 0 y 100 %** de que el cliente se vaya.

Eso es, en palabras, la **regresión logística**: un modelo que aprende cuántos puntos vale cada pista mirando la historia, y que convierte la suma en una probabilidad con una curva en forma de S (así nunca da menos de 0 ni más de 100).

<div style="border-left:3px solid #1B7E9C;background:#E1F0F5;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#14657D;">CÓMO SE LEE</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">Para que el modelo pueda sumar, las categorías se convierten en columnas de sí/no (contrato mes a mes: sí o no; fibra: sí o no) y los números se ponen en la misma escala. Eso lo hace la «preparación» de la celda; no hay que entenderla por dentro, solo saber que existe.</div></div>

```python
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preparar = ColumnTransformer([("num", StandardScaler(), NUMERICAS),
                              ("cat", OneHotEncoder(handle_unknown="ignore"), CATEGORICAS)])
logistica = Pipeline([("preparar", preparar), ("modelo", LogisticRegression(max_iter=2000))])
logistica.fit(X_entrenar, y_entrenar)

riesgo = logistica.predict_proba(X_probar)[:, 1]   # probabilidad de irse, cliente por cliente
prueba = X_probar.copy()
prueba["se_fue"] = y_probar.values
prueba["riesgo"] = riesgo
print("Modelo entrenado. Cinco clientes de prueba con su puntaje de riesgo:")
tabla(prueba[["tenure", "Contract", "InternetService", "MonthlyCharges", "riesgo", "se_fue"]].head(5)
      .rename(columns={"tenure": "Antigüedad", "Contract": "Contrato", "InternetService": "Internet",
                       "MonthlyCharges": "Cargo", "riesgo": "Riesgo", "se_fue": "¿Se fue?"}),
      formato={"Riesgo": "{:.0%}", "Cargo": "{:.0f} USD"})
```

## ¿Cuántos puntos vale cada pista?

Los **pesos** del modelo dicen cuánto suma o resta cada pista. Mostramos solo las que se entienden solas (de cada sí/no, una cara).

<div style="border-left:3px solid #1B7E9C;background:#E1F0F5;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#14657D;">CÓMO SE LEE</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">Barra hacia la derecha (roja): esa pista <b>suma riesgo</b>. Hacia la izquierda (azul): <b>resta</b>. El largo dice cuánto pesa. Es la tarjeta de puntos completa de esta telco.</div></div>

```python
NOMBRES = {"num__tenure": "Antigüedad (meses)", "num__MonthlyCharges": "Cargo mensual alto",
           "cat__Contract_Month-to-month": "Contrato mes a mes", "cat__Contract_Two year": "Contrato a dos años",
           "cat__InternetService_Fiber optic": "Internet por fibra óptica",
           "cat__PaymentMethod_Electronic check": "Paga con cheque electrónico",
           "cat__TechSupport_Yes": "Tiene soporte técnico", "cat__OnlineSecurity_Yes": "Tiene seguridad en línea",
           "cat__Dependents_Yes": "Tiene personas a cargo"}
nombres = logistica["preparar"].get_feature_names_out()
pesos = pd.Series(logistica["modelo"].coef_[0], index=nombres)
pesos = pesos[list(NOMBRES)].rename(NOMBRES).sort_values()

fig, ax = plt.subplots(figsize=(10, 4.8))
ax.barh(pesos.index, pesos.values, color=[ROJO if v > 0 else TEAL for v in pesos.values])
ax.axvline(0, color=TINTA, lw=1)
for i, v in enumerate(pesos.values):
    ax.text(v + (0.03 if v > 0 else -0.03), i, "suma" if v > 0 else "resta", va="center",
            ha="left" if v > 0 else "right", fontsize=9.5, color=GRIS)
ax.set_xticks([]); ax.grid(False)
encabezado(fig, "Figura 2 · La tarjeta de puntos",
           "El contrato mes a mes y el cargo alto suman riesgo; la antigüedad y el contrato largo lo restan",
           "Pesos de la regresión logística entrenada con el 80 % de los clientes")
plt.show()
```

## De los puntos a la probabilidad: la curva en S

La suma de puntos puede dar cualquier número: muy negativo (cliente tranquilo) o muy positivo (cliente en riesgo). La **curva en S** la convierte en una probabilidad: los muy negativos quedan cerca de 0 %, los muy positivos cerca de 100 %, y los del medio, en el medio.

<div style="border-left:3px solid #1B7E9C;background:#E1F0F5;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#14657D;">CÓMO SE LEE</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">No hace falta la fórmula. Lo que importa: <b>el 50 % está en el centro de la S</b>, y por eso el umbral «natural» de un modelo es 50 %. Pero ya veremos que el negocio casi nunca quiere ese umbral.</div></div>

```python
puntos = logistica.decision_function(X_probar)   # la suma de puntos de cada cliente de prueba
malla = np.linspace(-5, 5, 200)
fig, ax = plt.subplots(figsize=(10, 4.4))
ax.plot(malla, 1 / (1 + np.exp(-malla)), color=MAGENTA, lw=3)
ax.scatter(puntos, riesgo, s=10, color=GRIS, alpha=0.35, zorder=3)
ax.axhline(0.5, color=AMBAR, ls="--", lw=1.5); ax.axvline(0, color=AMBAR, ls="--", lw=1.5)
ax.text(0.15, 0.53, "50 %: el centro de la S", color=AMBAR, fontsize=10)
ax.set_xlabel("Suma de puntos de riesgo"); ax.set_ylabel("Probabilidad de irse")
ax.set_yticks([0, 0.25, 0.5, 0.75, 1]); ax.set_yticklabels(["0 %", "25 %", "50 %", "75 %", "100 %"])
encabezado(fig, "Figura 3 · La curva en S",
           "Cada punto gris es un cliente de prueba: su suma de puntos y la probabilidad que le toca",
           "Función logística; 1.407 clientes del 20 % de prueba")
plt.show()
```

# La matriz de confusión es el marcador del juego

Con un umbral (digamos 50 %), el modelo **decide**: por encima, «se va» (lo llamo); por debajo, «se queda» (no lo llamo). Comparado con lo que pasó de verdad, cada cliente cae en una de **cuatro casillas**. Son exactamente las filas de la hoja de marcador:

<div style="display:flex;gap:12px;flex-wrap:wrap;margin:12px 0 16px;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><div style="flex:1;min-width:180px;border-top:4px solid #2E8B57;background:#E3F2E9;padding:12px 14px;border-radius:0 0 8px 8px;"><div style="font-weight:700;color:#16201F;margin-bottom:4px;">Acierto</div><div style="font-size:10.5pt;line-height:1.5;color:#2A2933;">Lo llamé y se iba. Pagué 2 meses y conservé el año.</div></div><div style="flex:1;min-width:180px;border-top:4px solid #E8943A;background:#FCE9D3;padding:12px 14px;border-radius:0 0 8px 8px;"><div style="font-weight:700;color:#16201F;margin-bottom:4px;">Falsa alarma</div><div style="font-size:10.5pt;line-height:1.5;color:#2A2933;">Lo llamé y se quedaba igual. Pagué 2 meses de más.</div></div><div style="flex:1;min-width:180px;border-top:4px solid #C0392B;background:#FBE3E0;padding:12px 14px;border-radius:0 0 8px 8px;"><div style="font-weight:700;color:#16201F;margin-bottom:4px;">Fuga que no vi</div><div style="font-size:10.5pt;line-height:1.5;color:#2A2933;">No lo llamé y se fue. Perdí 12 meses.</div></div><div style="flex:1;min-width:180px;border-top:4px solid #1B7E9C;background:#E1F0F5;padding:12px 14px;border-radius:0 0 8px 8px;"><div style="font-weight:700;color:#16201F;margin-bottom:4px;">Tranquilidad</div><div style="font-size:10.5pt;line-height:1.5;color:#2A2933;">No lo llamé y se quedó. Cero.</div></div></div>

<div style="border-left:3px solid #2E8B57;background:#E3F2E9;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#1F6B42;">EN PLATA</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">Cada casilla tiene precio y el precio depende del cliente: perder a uno de 100 USD al mes duele más que perder a uno de 20. Sumando las casillas sale <b>el costo del mes</b>: lo que pagamos en ofertas más lo que perdimos en fugas que no vimos. <b>Menos es mejor.</b> Por eso vamos a contar <b>clientes</b> y <b>dólares</b> en cada casilla.</div></div>

```python
MESES_OFERTA, MESES_PERDIDA = 2, 12

def marcador(df_, columna_riesgo, umbral):
    """Cuenta clientes y dólares en las cuatro casillas para un umbral dado."""
    llamo = df_[columna_riesgo] >= umbral
    se_fue = df_["se_fue"] == 1
    cargo = df_["MonthlyCharges"]
    acierto, falsa, fuga, tranq = llamo & se_fue, llamo & ~se_fue, ~llamo & se_fue, ~llamo & ~se_fue
    return {"umbral": umbral, "llamadas": int(llamo.sum()),
            "aciertos": int(acierto.sum()), "falsas_alarmas": int(falsa.sum()),
            "fugas_no_vistas": int(fuga.sum()), "tranquilidades": int(tranq.sum()),
            "ofertas": float((MESES_OFERTA * cargo[llamo]).sum()),
            "fugas": float((MESES_PERDIDA * cargo[fuga]).sum())}

def total(m):
    """Costo del mes: ofertas pagadas + fugas que no vimos. Menos es mejor."""
    return m["ofertas"] + m["fugas"]

m50 = marcador(prueba, "riesgo", 0.5)
fig, ax = plt.subplots(figsize=(8.5, 5))
cargo_p = prueba["MonthlyCharges"]; llamo50 = prueba.riesgo >= 0.5; sefue = prueba.se_fue == 1
celdas = [[("Acierto", m50["aciertos"], (MESES_OFERTA * cargo_p[llamo50 & sefue]).sum(), VERDE),
           ("Falsa alarma", m50["falsas_alarmas"], (MESES_OFERTA * cargo_p[llamo50 & ~sefue]).sum(), AMBAR)],
          [("Fuga que no vi", m50["fugas_no_vistas"], m50["fugas"], ROJO), ("Tranquilidad", m50["tranquilidades"], 0, TEAL)]]
for i, fila in enumerate(celdas):
    for j, (nombre, n, plata, color) in enumerate(fila):
        ax.add_patch(plt.Rectangle((j, 1 - i), 1, 1, color=color, alpha=0.18))
        ax.text(j + 0.5, 1 - i + 0.68, nombre, ha="center", fontsize=12, fontweight="bold", color=TINTA)
        ax.text(j + 0.5, 1 - i + 0.42, f"{n} clientes", ha="center", fontsize=12, color=CUERPO)
        ax.text(j + 0.5, 1 - i + 0.18, f"cuesta {plata:,.0f} USD".replace(",", "."), ha="center", fontsize=11.5, color=color, fontweight="bold")
ax.set_xlim(0, 2); ax.set_ylim(0, 2); ax.set_xticks([0.5, 1.5]); ax.set_xticklabels(["Se iba", "Se quedaba"], fontsize=11)
ax.set_yticks([1.5, 0.5]); ax.set_yticklabels(["Lo llamé", "No lo llamé"], fontsize=11); ax.grid(False)
for s in ax.spines.values(): s.set_visible(False)
encabezado(fig, "Figura 4 · La matriz con precio",
           f"Con umbral 50 %, el mes cuesta {total(m50):,.0f} USD en el 20 % de prueba".replace(",", "."),
           "Clientes y costo en cada casilla; oferta = 2 meses del cargo, fuga = 12 meses; sin modelo costaría " + f"{(MESES_PERDIDA * cargo_p[sefue]).sum():,.0f} USD".replace(",", "."))
plt.show()
```

## La trampa de la exactitud

La **exactitud** es el porcentaje de clientes que el modelo clasificó bien (aciertos + tranquilidades). Suena razonable, y en churn engaña. Mira:

```python
exactitud_modelo = (m50["aciertos"] + m50["tranquilidades"]) / len(prueba)
exactitud_nadie = 1 - prueba.se_fue.mean()
print(f"Exactitud del modelo con umbral 50 %:        {exactitud_modelo:.1%}")
print(f"Exactitud de decir «nadie se va» (sin modelo): {exactitud_nadie:.1%}")
print(f"Clientes que salva «nadie se va»:              0")
```

## Las dos preguntas que sí sirven

Quien decide cuál pesa más es <b>el presupuesto</b>, no el modelo.</div></div>

```python
def precision_cobertura(m):
    p = m["aciertos"] / m["llamadas"] if m["llamadas"] else 0
    c = m["aciertos"] / (m["aciertos"] + m["fugas_no_vistas"])
    return p, c

p, c = precision_cobertura(m50)
print(f"Con umbral 50 %: precisión {p:.0%} (de cada 100 llamadas, {p*100:.0f} eran fugas reales)")
print(f"                 cobertura {c:.0%} (de cada 100 que se iban, alcanzamos {c*100:.0f})")
```

# ¿Desde qué puntaje llamo? El umbral lo pone el negocio

El modelo entrega un puntaje; **el umbral lo decide la empresa**. Bajarlo es llamar a más gente: más aciertos y más falsas alarmas. Subirlo es ser más selectivo. Vamos a probar todos los umbrales de 5 % a 95 % y a ver qué pasa con las llamadas, las dos métricas y, sobre todo, el costo del mes.

Antes de correrlo, piensa la cuenta: la oferta cuesta 2 meses y la fuga 12. Llamar a un cliente vale la pena cuando su riesgo pasa de **2 entre 12, uno de cada seis (17 %)**. Si el modelo está bien calibrado, el mejor umbral debería andar por ahí.

<div style="border-left:3px solid #8E5BD9;background:#F0E7FA;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#6B3FB0;">EJECUTA</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">Si quieres jugar: cambia <code>MESES_OFERTA</code> o <code>MESES_PERDIDA</code> arriba y vuelve a correr desde la celda de la matriz. # El árbol de decisión: preguntas de sí o no

La tarjeta de puntos suma. El **árbol de decisión** pregunta. Empieza por la pregunta que mejor separa a los que se van de los que se quedan («¿contrato mes a mes?»), y en cada rama vuelve a preguntar. Al final de cada camino hay una hoja con el porcentaje de clientes que se fueron por ahí: ese es el puntaje de riesgo del árbol.

Lo limitamos a **tres preguntas de profundidad** para que quepa en una lámina y se pueda leer como reglas de negocio.

<div style="border-left:3px solid #1B7E9C;background:#E1F0F5;border-radius:0 10px 10px 0;padding:13px 20px;margin:12px 0;font-family:'IBM Plex Sans', system-ui, -apple-system, 'Segoe UI', Roboto, sans-serif;"><span style="font-family:'IBM Plex Mono', ui-monospace, Menlo, monospace;font-size:8.5pt;letter-spacing:2px;color:#14657D;">CÓMO SE LEE</span><div style="font-size:11.5pt;line-height:1.6;color:#2A2933;">En cada caja: la pregunta, cuántos clientes de entrenamiento llegaron ahí y cuántos se fueron. Cuanto más rojo, más riesgo. Lee un camino completo de arriba abajo: es una regla que cualquier persona de la empresa puede aplicar sin computador.</div></div>

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree

preparar_arbol = ColumnTransformer([("num", "passthrough", NUMERICAS),
                                    ("cat", OneHotEncoder(handle_unknown="ignore"), CATEGORICAS)])
arbol = Pipeline([("preparar", preparar_arbol),
                  ("modelo", DecisionTreeClassifier(max_depth=3, min_samples_leaf=50, random_state=42))])
arbol.fit(X_entrenar, y_entrenar)
prueba["riesgo_arbol"] = arbol.predict_proba(X_probar)[:, 1]

ETIQUETAS = {"num__tenure": "antigüedad (meses)", "num__MonthlyCharges": "cargo mensual",
             "cat__Contract_Month-to-month": "contrato mes a mes", "cat__Contract_One year": "contrato a un año",
             "cat__InternetService_Fiber optic": "fibra óptica"}
nombres_arbol = [ETIQUETAS.get(n, n.split("__")[-1]) for n in arbol["preparar"].get_feature_names_out()]
fig, ax = plt.subplots(figsize=(14, 6.5))
plot_tree(arbol["modelo"], feature_names=nombres_arbol, class_names=["se queda", "se va"], filled=True,
          impurity=False, proportion=False, rounded=True, fontsize=9, ax=ax, label="root")
encabezado(fig, "Figura 6 · El árbol en tres preguntas",
           "Contrato mes a mes, fibra óptica y poca antigüedad: el camino más peligroso",
           "Árbol de decisión de profundidad 3 entrenado con el 80 %; en cada caja, cuántos se quedaron y cuántos se fueron")
plt.show()
```

# Del puntaje al plan de retención

Un modelo de churn se usa así, cada mes:

1. Se **calcula el riesgo** de todos los clientes activos con el modelo entrenado.
2. Se **ordenan** de mayor a menor riesgo.
3. El **presupuesto** dice hasta dónde llega la lista (ese es el umbral).
4. Se llama, se registra qué pasó y, con esa historia nueva, el modelo se **reentrena**.

Así se puntúa un cliente nuevo, por ejemplo uno que acaba de firmar un contrato mes a mes con fibra:
